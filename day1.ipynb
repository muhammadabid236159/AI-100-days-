## Arithmetic Operators

Basic math operations: +, -, *, /, //, %, **

```python
a = 5
b = 7
sum_two_numbers = a + b
sum_two_numbers
```

```python
print(f"Addition: {a} + {b} = {a+b}")
print(f"Subtraction: {a} - {b} = {a-b}")
print(f"Multiplication: {a} * {b} = {a*b}")
print(f"Division: {a} / {b} = {a/b}")
print(f"Modulus: {a} % {b} = {a%b}")
print(f"Exponent: {a} ** 2 = {a**2}")
```

## Relational Operators

Compare values: >, <, >=, <=, !=, ==

```python
print(f"5 > 7? {a > b}")
print(f"5 < 7? {a < b}")
print(f"5 >= 7? {a >= b}")
print(f"5 <= 7? {a <= b}")
print(f"5 != 7? {a != b}")
print(f"5 == 7? {a == b}")
```

## Assignment Operators

Assign values: =, +=, -=, *=, /=, etc.

```python
x = 10
print(f"Initial: x = {x}")

x += 5  # x = x + 5
print(f"After += 5: x = {x}")

x -= 3  # x = x - 3
print(f"After -= 3: x = {x}")

x *= 2  # x = x * 2
print(f"After *= 2: x = {x}")
```

## Logical Operators

Boolean logic: and, or, not

### Truth Tables

| A | B | A and B | A or B | not A |
|---|---|---------|--------|-------|
| 0 | 0 |    0    |   0    |   1   |
| 0 | 1 |    0    |   1    |   1   |
| 1 | 0 |    0    |   1    |   0   |
| 1 | 1 |    1    |   1    |   0   |

```python
print(f"5 and 7: {bool(5 and 7)}")
print(f"0 or 1: {bool(0 or 1)}")
print(f"not 5: {not 5}")
```

## Operator Precedence

Priority order (highest to lowest):

| Priority | Operator | Description |
|----------|----------|-------------|
| 1 | () | Parentheses |
| 2 | ** | Exponentiation |
| 3 | +x, -x | Unary plus, minus |
| 4 | *, /, //, % | Multiplication, division |
| 5 | +, - | Addition, subtraction |
| 6 | ==, !=, >, <, >=, <= | Comparisons |
| 7 | not | Logical NOT |
| 8 | and | Logical AND |
| 9 | or | Logical OR |

## Type Conversion

Converting between data types:
- **Implicit:** Python does it automatically
- **Explicit:** We do it manually (int(), str(), float())

```python
# Type checking
name = "abid"
print(f"Type of 'abid': {type(name)}")

number = 42
print(f"Type of 42: {type(number)}")

decimal = 3.14
print(f"Type of 3.14: {type(decimal)}")
```
