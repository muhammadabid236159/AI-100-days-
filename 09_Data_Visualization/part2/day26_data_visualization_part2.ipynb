```python
import matplotlib.pyplot as plt
import numpy as np

# 1. 100 matches ka score generate karein
scores = np.random.normal(loc=50, scale=15, size=100)

# 2. Histogram banayein
plt.hist(scores, bins=10, edgecolor="black", color="orange")
plt.xlabel("Runs")
plt.ylabel("Number of Matches (Count)")
plt.title("Batsman Scores Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Do alag alag datasets ke data generate karein
section_a_marks = np.random.normal(70, 10, 100)
section_b_marks = np.random.normal(65, 12, 100)

# 2. Histograms plot karein (alpha=0.5 se transparency aati hai)
plt.hist(
    section_a_marks,
    bins=10,
    alpha=0.5,
    label="Section A",
    color="blue",
    edgecolor="black",
)
plt.hist(
    section_b_marks,
    bins=10,
    alpha=0.5,
    label="Section B",
    color="orange",
    edgecolor="black",
)

# 3. Labels, Title aur Legend jo aapne mention kiye thay
plt.xlabel("Marks / Scores")
plt.ylabel("Number of Students (Frequency)")
plt.title("Comparing Two Sections (Overlay Histogram)")
plt.legend()

# 4. Graph ko screen par dikhane ke liye
plt.show()
```

```python
# Box Plot
data = {
    "CS": np.random.normal(70, 10, 100),
    "SE": np.random.normal(65, 12, 100),
    "AI": np.random.normal(80, 8, 100),
    "DS": np.random.normal(75, 15, 100)
}

df = pd.DataFrame(data)

plt.figure(figsize=(10, 5))
plt.boxplot(df.values, labels=df.columns, patch_artist=True,
            boxprops=dict(facecolor="skyblue"),
            medianprops=dict(color="red", linewidth=2))
plt.title("Department Wise Marks Distribution", fontsize=14, fontweight="bold")
plt.xlabel("Department")
plt.ylabel("Marks")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"
]

cities = ["Mingora", "Peshawar", "Islamabad", "Lahore"]

temperature = [
    [28, 31, 29, 30, 32, 33, 31],
    [35, 36, 37, 38, 39, 40, 38],
    [27, 28, 29, 30, 28, 29, 27],
    [34, 35, 36, 37, 38, 39, 40]
]

fig, axes = plt.subplots(2, 2, figsize=(10, 7))

for i in range(2):
    for j in range(2):

        index = i * 2 + j

        axes[i][j].plot(days, temperature[index])
        axes[i][j].set_title(cities[index])

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 5))

axes[0][0].plot(days, temperature[0])
axes[0][0].set_title("Mingora")

axes[0][1].plot(days, temperature[1])
axes[0][1].set_title("Peshawar")

axes[1][0].plot(days, temperature[2])
axes[1][0].set_title("Islamabad")

axes[1][1].plot(days, temperature[3])
axes[1][1].set_title("Lahore")

plt.tight_layout()
plt.show()
```
