```python
import numpy as np
import matplotlib.pyplot as plt

# Random data
years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
oscar_revenue = np.random.randint(100, 800, size=10)
non_oscar_revenue = np.random.randint(500, 2500, size=10)

# Simple comparison
plt.figure(figsize=(12, 5))
plt.plot(years, oscar_revenue, marker="o", color="gold", linewidth=2, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, marker="s", color="red", linewidth=2, label="Non-Oscar Movies")
plt.title("Oscar vs Non-Oscar Revenue Comparison", fontsize=14, fontweight="bold")
plt.xlabel("Year")
plt.ylabel("Revenue (Million USD)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]

cs_marks = [68, 72, 75, 70, 78, 74, 81, 85, 83, 88]

se_marks = [65, 69, 73, 76, 74, 79, 80, 82, 86, 84]

plt.figure(figsize=(12, 6))

plt.title("Computer Science vs Software Engineering Marks")

plt.plot(years, cs_marks, marker="o", linewidth="1", color="gold",linestyle='-',label="Computer Science")
plt.plot(years, se_marks, marker="s" , color="blue", linestyle="--", label="Software Engineering")

plt.xlabel("Years")
plt.ylabel("Average Marks")
plt.grid()
plt.legend(loc="right")
plt.tight_layout()
plt.savefig("img.png")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
cs_marks = [68, 72, 75, 70, 78, 74, 81, 85, 83, 88]
se_marks  = [65, 69, 73, 76, 74, 79, 80, 82, 86, 84]

x = np.arange(len(years))
width = 0.4

plt.figure(figsize=(12, 6))

plt.bar(x + width/2, cs_marks, width=width, color="skyblue", label="CS", edgecolor="black")
plt.bar(x - width/2, se_marks, width=width, color="black",  label="SE", edgecolor="black")

plt.title("SE vs CS Marks (2015-2024)", fontsize=14, fontweight="bold")
plt.xlabel("Year", fontsize=12)
plt.ylabel("Marks", fontsize=12)
plt.xticks(x, years)
plt.yticks(range(0, 100, 10))
plt.legend(fontsize=12)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]
cs_marks = [68, 72, 75, 70, 78, 74, 81, 85, 83, 88]
se_marks  = [65, 69, 73, 76, 74, 79, 80, 82, 86, 84]

x = np.arange(len(years))
width = 0.4

plt.figure(figsize=(10, 8))

plt.barh(x + width/2, cs_marks, height=width, color="skyblue", label="CS", edgecolor="black")
plt.barh(x - width/2, se_marks, height=width, color="salmon",  label="SE", edgecolor="black")

plt.title("SE vs CS Marks (2015-2024)", fontsize=14, fontweight="bold")
plt.xlabel("Marks", fontsize=12)
plt.ylabel("Year", fontsize=12)
plt.yticks(x, years)
plt.legend(fontsize=12)
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 2.5, 3, 4, 4.5, 5, 6, 7, 8, 8, 1]
marks = [45, 50, 55, 58, 65, 68, 72, 78, 85, 90, 20, 95]
colors = ["green" if mark >= 50 else "red" for mark in marks]
plt.scatter(
    study_hours,
    marks,
   marker="o",
   cmap="cividis",
   c=["darkblue"]*len(marks)
)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

plt.grid(True, linestyle="--", alpha=0.4)
plt.colorbar()
plt.show()
```
