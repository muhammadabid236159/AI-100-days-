## If-Elif-Else Statements

Conditional statements based on true/false values

```python
age = 20
if age > 18:
    print("you can vote")
elif age == 18:
    print("you are 18, just became eligible")
else:
    print("you cannot vote yet")
```

```python
age = 25  # Can be taken from user input

if age <= 13:
    print("child")
elif age > 13 and age <= 18:
    print("teenager")
else:
    print("adult")
```

## Nested If-Else

If statements inside if statements for complex logic

```python
username = "Admin"
password = "12345"

if username == "Admin" and password == "12334":
    print("login successful")
else:
    if username != "Admin":
        print("username is incorrect")
    else:
        print("wrong password")
```

## Loops in Python

Repeat code multiple times

### While Loop

```python
counter = 1
while counter <= 5:
    print("hello")
    counter += 1
```

### For Loop with Range

```python
for n in range(1, 5, 2):  # start, end, step
    print(n)
```

## 🏥 PROJECT: Patient Management System

A complete CLI system for hospital management
- Register patients
- View patient list
- Book appointments with doctors

**Concepts Used:**
- While loops (menu system)
- Dictionaries (patient storage)
- Lists (collections)
- String formatting (f-strings)
- User input validation

```python
print("🏥 WELCOME TO CLINIC")

patients = []
patient_id = 1

doctors = [
    "Dr. Ahmed (Cardiologist)",
    "Dr. Sara (Dermatologist)",
    "Dr. Ali (General Physician)"
]

while True:
    print("\n--- MENU ---")
    print("1. Register Patient")
    print("2. View Patients")
    print("3. Appointment Booking")
    print("4. Exit")

    choice = input("Enter choice: ")

    # 1. REGISTER
    if choice == "1":
        name = input("Enter Name: ")
        age = input("Enter Age: ")
        disease = input("Enter Disease: ")

        if name == "" or age == "" or disease == "":
            print("❌ Please fill all fields")
            continue

        patient = {
            "id": patient_id,
            "name": name,
            "age": age,
            "disease": disease
        }

        patients.append(patient)
        print(f"✅ Patient Registered! ID = {patient_id}")
        patient_id += 1

    # 2. VIEW PATIENTS
    elif choice == "2":
        if len(patients) == 0:
            print("⚠️ No patients registered yet")
        else:
            print("--- PATIENT LIST ---")
            for p in patients:
                print(f"ID: {p['id']}, Name: {p['name']}, Age: {p['age']}, Disease: {p['disease']}")

    # 3. APPOINTMENT BOOKING
    elif choice == "3":
        print("\n--- DOCTORS LIST ---")

        i = 1
        for doctor in doctors:
            print(f"{i}. {doctor}")
            i += 1

        doc_choice = int(input("Select doctor (1-3): "))

        if 1 <= doc_choice <= len(doctors):
            print(f"✅ Appointment booked with {doctors[doc_choice - 1]}")
        else:
            print("❌ Invalid doctor selection")

    # 4. EXIT
    elif choice == "4":
        print("Exiting system...")
        break

    else:
        print("❌ Invalid choice")
```
