```python

import mysql.connector

db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",
    database="ai_ml_db"
)

print("Database connected")

query = """
SELECT 
    students.id,
    students.name AS student_name,
    students.age,
    students.marks,
    students.city,
    teacher.name AS teacher_name
FROM students
INNER JOIN teacher
ON students.id = teacher.s_id
"""

df = pd.read_sql(query, db)

print(df)


```

```python
query = """
SELECT *
FROM students
JOIN teacher
ON students.id = teacher.s_id
"""

df = pd.read_sql(query, db)

print(df)
```

```python


query_teacher = """
SELECT 
    teacher.id,
    teacher.name,
    classes.class_name
FROM teacher
INNER JOIN classes
ON teacher.id = classes.teacher_id
WHERE teacher.id=1

"""

teacher = pd.read_sql(query_teacher, db)

print(teacher)
```
