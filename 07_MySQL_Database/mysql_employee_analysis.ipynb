```python
query="SELECT EmpID, FirstName, LastName,salary FROM employees"
newdata=pd.read_sql(query, db)
print(newdata)

newdata.set_index('EmpID', inplace=True)
query="SELECT * FROM employees where department='IT' AND salary>6000"
it_employees=pd.read_sql(query, db)
print(it_employees)
```

```python
query = """
SELECT EmpID, FirstName, LastName, Department, Salary, HireDate
FROM employees
ORDER BY HireDate DESC
"""

newdata = pd.read_sql(query, db, index_col="EmpID")

print(newdata)
```
