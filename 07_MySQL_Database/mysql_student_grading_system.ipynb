```python
import mysql.connector
import numpy as np
import pandas as pd

db=mysql.connector.connect(
    host="localhost",
        user="root",
        password="",
        database="school_db"
)
print("database cannected")
```

```python
query="SELECT * FROM students"
df=pd.read_sql(query,db)
df.head()
df.info()
df
```

```python
df["total"]=df['subject1']+df['subject2']+df['subject3']
max_marks=300
df['percentage']=(df['total']/max_marks)*100
df['percentage']=df['percentage'].round(2)
```

```python

import numpy as np
def get_grade(percentage):
    if percentage >= 90:
        return 'A'
    elif percentage >= 80:
        return 'B'
    elif percentage >= 70:
        return 'C'
    elif percentage >= 60:
        return 'D'
    elif percentage >= 40:
        return 'E'
    else:
        return 'Fail'
    
```

```python
df['grade']=df['percentage'].apply(get_grade)
percentages =df['percentage'].values
class_average = np.mean(percentages)
print("Class Average:", round(class_average, 2))
highest = np.max(percentages)
lowest = np.min(percentages)
print("Highest:", highest)
print("Lowest:", lowest)
std_dev = np.std(percentages)
print("Standard Deviation:", round(std_dev, 2))
topper_index = np.argmax(percentages)   # highest value ka index
topper = df.iloc[topper_index]
print("Topper:", topper['name'], "-", topper['percentage'])
df
```
