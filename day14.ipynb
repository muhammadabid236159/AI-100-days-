```python
import numpy as np

area = np.array([1000, 1200, 1500, 1800, 2000])

mean = np.mean(area)
std = np.std(area)

normalized_area = (area - mean) / std

print("Mean:", mean)
print("Std:", std)
print("Standardized Area:", normalized_area)
```
