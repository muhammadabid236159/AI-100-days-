## 📚 1. Probability - Theory & Basics

**Probability** = Kisi event ke hone ka chance (0 se 1 ke beech).

### Formula:
$$P(A) = \frac{\text{Favorable Outcomes}}{\text{Total Outcomes}}$$

### Rules:
- **P(A) = 0** $\rightarrow$ Impossible Event (kabhi nahi hoga)
- **P(A) = 1** $\rightarrow$ Sure Event (zaroor hoga)
- **0 ≤ P(A) ≤ 1** $\rightarrow$ Har probability 0 aur 1 ke beech hoti hai
- **P(A) + P(A') = 1** $\rightarrow$ Event + Complement = 1

```python
# 🎲 Probability Basics - Code Implementation

# 1. Dice Roll Sample Space
sample_space_dice = {1, 2, 3, 4, 5, 6}
total_dice = len(sample_space_dice)

event_4 = {4}
event_even = {2, 4, 6}
event_gt4 = {5, 6}

print(f"🎲 P(getting 4)        = {len(event_4)}/{total_dice} = {len(event_4)/total_dice:.4f}")
print(f"🎲 P(even number)     = {len(event_even)}/{total_dice} = {len(event_even)/total_dice:.4f}")
print(f"🎲 P(greater than 4)  = {len(event_gt4)}/{total_dice} = {len(event_gt4)/total_dice:.4f}")

# 2. Coin Toss
coin_space = {"Head", "Tail"}
print(f"\n🪙 P(Head)             = 1/{len(coin_space)} = {1/len(coin_space):.4f}")
print(f"🪙 P(Tail)             = 1/{len(coin_space)} = {1/len(coin_space):.4f}")

# 3. Card Draw (52 Cards)
total_cards = 52
hearts = 13
kings = 4
print(f"\n🃏 P(Heart)            = {hearts}/{total_cards} = {hearts/total_cards:.4f}")
print(f"🃏 P(King)             = {kings}/{total_cards} = {kings/total_cards:.4f}")

# Complement Event: P(A') = 1 - P(A)
p_not_king = 1 - (kings / total_cards)
print(f"🃏 P(Not King)         = 1 - {kings/total_cards:.4f} = {p_not_king:.4f}")
```

## 📚 2. Types of Events - Theory

1. **Sure Event (Certain Event):** Zaroor hoga $\rightarrow P(A) = 1$ (e.g., Dice par number $\le 6$ aana).
2. **Impossible Event:** Kabhi nahi ho sakta $\rightarrow P(A) = 0$ (e.g., Dice par 7 aana).
3. **Equally Likely Events:** Sab outcomes ka barabar chance ho (e.g., Fair coin toss).
4. **Complementary Events:** Event ka na hona $\rightarrow P(A) + P(A') = 1$.
5. **Mutually Exclusive Events:** Dono events ek saath nahi ho sakte $\rightarrow P(A \cap B) = 0$.
6. **Independent Events:** Ek event dusre ko affect na kare $\rightarrow P(A \cap B) = P(A) \times P(B)$.
7. **Dependent Events:** Ek event ke hone se dusre ki probability badal jaye.

```python
# 📌 Types of Events - Code Implementation

dice = {1, 2, 3, 4, 5, 6}
total_outcomes = len(dice)

# 1️⃣ Sure Event (P = 1)
less_than_7 = {1, 2, 3, 4, 5, 6}
p_sure = len(less_than_7) / total_outcomes
print(f"1️⃣ Sure Event: P(number < 7) = {p_sure:.2f} ✅")

# 2️⃣ Impossible Event (P = 0)
equal_to_7 = set()
p_impossible = len(equal_to_7) / total_outcomes
print(f"2️⃣ Impossible Event: P(number = 7) = {p_impossible:.2f} ❌")

# 3️⃣ Equally Likely Events
coin = {"Head", "Tail"}
print(f"3️⃣ Equally Likely: P(Head) = {1/len(coin):.2f}, P(Tail) = {1/len(coin):.2f}")

# 4️⃣ Complementary Events: P(A) + P(A') = 1
event_even = {2, 4, 6}
p_even = len(event_even) / total_outcomes
p_odd = 1 - p_even
print(f"4️⃣ Complementary: P(Even) = {p_even:.2f}, P(Odd) = {p_odd:.2f} (Sum = {p_even + p_odd:.2f})")

# 5️⃣ Mutually Exclusive Events: P(A ∩ B) = 0
event_3 = {3}
event_5 = {5}
intersection = event_3 & event_5
print(f"5️⃣ Mutually Exclusive: P(3 ∩ 5) = {len(intersection)/total_outcomes:.2f}")

# 6️⃣ Independent Events: P(A ∩ B) = P(A) * P(B)
p_head1 = 0.5
p_head2 = 0.5
p_both_heads = p_head1 * p_head2
print(f"6️⃣ Independent: P(Head₁ ∩ Head₂) = {p_head1} × {p_head2} = {p_both_heads:.2f}")

# 7️⃣ Dependent Events (Without replacement)
total_balls = 5  # 3 Red, 2 Blue
p_red1 = 3 / total_balls
p_red2_given_red1 = 2 / (total_balls - 1)
p_both_red = p_red1 * p_red2_given_red1
print(f"7️⃣ Dependent: P(Both Red) = {p_red1:.2f} × {p_red2_given_red1:.2f} = {p_both_red:.2f}")
```

## ⚽🏏 3. Union & Intersection Problem

**Problem Statement:**
- Total Students = 50
- Football ($A$) = 20
- Cricket ($B$) = 30
- Both ($A \cap B$) = 10

**Union Formula:**
$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

```python
# ⚽🏏 Student Sports Problem
total_students = 50
football = 20
cricket = 30
both = 10

# Calculations
intersection = both
union = football + cricket - both
only_football = football - both
only_cricket = cricket - both
no_sport = total_students - union

print(f"⚽🏏 Intersection (A ∩ B) = {intersection} students (dono khelty hain)")
print(f"🔵 Union (A ∪ B)        = {football} + {cricket} - {both} = {union} students (koi bhi sport)")
print(f"⚽ Sirf Football        = {only_football} students")
print(f"🏏 Sirf Cricket         = {only_cricket} students")
print(f"❌ Koi sport nahi       = {no_sport} students")

print("\n--- Probabilities ---")
print(f"P(Football)  = {football}/{total_students} = {football/total_students:.2f}")
print(f"P(Cricket)   = {cricket}/{total_students} = {cricket/total_students:.2f}")
print(f"P(Both)      = {both}/{total_students} = {both/total_students:.2f}")
print(f"P(A ∪ B)     = {union}/{total_students} = {union/total_students:.2f}")
print(f"P(No Sport)  = {no_sport}/{total_students} = {no_sport/total_students:.2f}")
```

## 🏭 4. Law of Total Probability - Theory & Formula

### Concept:
Agar poore Sample Space $S$ ko alag-alag mutually exclusive groups (partitions) $B_1, B_2, \dots, B_n$ mein baanta gaya ho, aur koi event $A$ in sab partitions se hokar guzarta ho:

Toh event $A$ ki total probability in tamam branches ke pieces ka sum hoti hai:

### Formula:
$$P(A) = \sum_{i=1}^{n} P(A \cap B_i) = \sum_{i=1}^{n} P(A|B_i) \cdot P(B_i)$$

**Example (Factory Production & Defect):**
- Machine 1 ($B_1$): 50% items banati hai $\rightarrow P(B_1) = 0.50$, Defect Rate $P(D|B_1) = 2\% = 0.02$
- Machine 2 ($B_2$): 30% items banati hai $\rightarrow P(B_2) = 0.30$, Defect Rate $P(D|B_2) = 3\% = 0.03$
- Machine 3 ($B_3$): 20% items banati hai $\rightarrow P(B_3) = 0.20$, Defect Rate $P(D|B_3) = 5\% = 0.05$

**Total Defective Item ($D$) nikalne ki probability:**
$$P(D) = P(D|B_1)P(B_1) + P(D|B_2)P(B_2) + P(D|B_3)P(B_3)$$

```python
# 🏭 Law of Total Probability - Code Implementation

# Machine wise production shares (Priors)
p_b1 = 0.50  # Machine 1 makes 50% items
p_b2 = 0.30  # Machine 2 makes 30% items
p_b3 = 0.20  # Machine 3 makes 20% items

# Machine wise defect rates (Conditionals: P(Defect | Machine))
p_d_given_b1 = 0.02  # 2% defect rate
p_d_given_b2 = 0.03  # 3% defect rate
p_d_given_b3 = 0.05  # 5% defect rate

# Har machine ka defect contribution: P(D ∩ B_i) = P(D | B_i) * P(B_i)
p_defect_b1 = p_d_given_b1 * p_b1
p_defect_b2 = p_d_given_b2 * p_b2
p_defect_b3 = p_d_given_b3 * p_b3

# Law of Total Probability: P(Defect) = sum of all contributions
total_p_defect = p_defect_b1 + p_defect_b2 + p_defect_b3

print("--- 🏭 Law of Total Probability (Factory Defect Rate) ---")
print(f"Machine 1 Defect Contribution: {p_d_given_b1} × {p_b1} = {p_defect_b1:.4f}")
print(f"Machine 2 Defect Contribution: {p_d_given_b2} × {p_b2} = {p_defect_b2:.4f}")
print(f"Machine 3 Defect Contribution: {p_d_given_b3} × {p_b3} = {p_defect_b3:.4f}")
print(f"\n🔴 Total Probability of Defective Item P(Defect) = {total_p_defect:.4f} ({total_p_defect*100:.2f}%)")
```

## 🧠 5. Bayes' Theorem - Theory & Formula

### Concept:
Bayes' Theorem humein kisi event ki probability **update** karne ka tareeqa sikhata hai jab humein koi **naya evidence** mil jaye!

Machine Learning mein **Naive Bayes Classifier** (Spam Detection, Sentiment Analysis, Medical Diagnosis) ispar mubni hai.

### Formula:
$$P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$$

Partition form (Machine / Hypothesis testing):
$$P(B_k|A) = \frac{P(A|B_k) \cdot P(B_k)}{\sum_{i=1}^n P(A|B_i) \cdot P(B_i)} = \frac{P(A|B_k) \cdot P(B_k)}{P(A)}$$

### 4 Important Terms:
1. **Prior Probability $P(A)$:** Naya evidence aane se pehle ka yaqeen.
2. **Likelihood $P(B|A)$:** Agar hypothesis $A$ sach ho toh evidence $B$ dekhne ka chance.
3. **Marginal Probability / Evidence $P(B)$:** Total evidence ka chance (Total Law of Probability se).
4. **Posterior Probability $P(A|B)$:** Evidence milne ke baad updated probability.

$$\text{Posterior} = \frac{\text{Likelihood} \times \text{Prior}}{\text{Evidence}}$$

```python
# Example: Medical Test / ML Classifier (Disease Diagnosis)
# - Disease Prevalence P(Disease) = 1% = 0.01
# - Test Sensitivity (True Positive Rate) P(+ | Disease) = 99% = 0.99
# - False Positive Rate P(+ | Healthy) = 5% = 0.05
# Question: Agar kisi ka test Positive aaya, use waqai bimari hone ka kitna chance hai?

p_disease = 0.01
p_healthy = 1 - p_disease  # 0.99

p_pos_given_disease = 0.99  # True positive rate
p_pos_given_healthy = 0.05  # False positive rate

# Total probability of test being positive (Law of Total Probability)
p_positive = (p_pos_given_disease * p_disease) + (p_pos_given_healthy * p_healthy)

# Bayes' Theorem: P(Disease | Positive)
p_disease_given_pos = (p_pos_given_disease * p_disease) / p_positive

print("--- Medical Diagnosis / ML Classifier Example ---")
print(f"Prior P(Disease)                = {p_disease:.2%}")
print(f"Total Chance of Positive Test   = {p_positive:.4f} ({p_positive*100:.2f}%)")
print(f"🎯 Posterior P(Disease | Pos)    = {p_disease_given_pos:.4f} ({p_disease_given_pos*100:.2f}%)")
print("\nNateeja: 99% accurate test hone ke bawajood bimari ka chance sirf ~16.6% hai kyunki bimari bohat rare hai!")
```

```python
# 🧠 Bayes' Theorem - Factory Machine Problem (Self-Contained)

# ==========================================================
# 1. GIVEN VALUES
# ==========================================================
# Production shares (Priors)
p_m1 = 0.50  # Machine 1 makes 50%
p_m2 = 0.30  # Machine 2 makes 30%
p_m3 = 0.20  # Machine 3 makes 20%

# Defect rates (Likelihood)
p_d_given_m1 = 0.02  # Machine 1: 2% defect
p_d_given_m2 = 0.03  # Machine 2: 3% defect
p_d_given_m3 = 0.05  # Machine 3: 5% defect


# ==========================================================
# 2. STEP 1: Law of Total Probability -> Find P(Defect)
# ==========================================================
defect_from_m1 = p_m1 * p_d_given_m1
defect_from_m2 = p_m2 * p_d_given_m2
defect_from_m3 = p_m3 * p_d_given_m3

total_p_defect = defect_from_m1 + defect_from_m2 + defect_from_m3


# ==========================================================
# 3. STEP 2: Bayes' Theorem -> Find P(M3 | Defect)
# Formula: P(M3 | D) = [ P(D | M3) * P(M3) ] / P(D)
# ==========================================================
numerator = p_d_given_m3 * p_m3          # Likelihood * Prior
denominator = total_p_defect             # Total Evidence P(D)
p_m3_given_defect = numerator / denominator

print("=========== BAYES' THEOREM CALCULATION ===========")
print(f"Prior P(M3)                     = {p_m3:.2f} (20%)")
print(f"Likelihood P(D | M3)            = {p_d_given_m3:.2f} (5%)")
print(f"Total Defect Chance P(Defect)   = {total_p_defect:.4f} (2.9%)")
print(f"\n🎯 P(Machine 3 | Defect)        = {numerator:.4f} / {denominator:.4f}")
print(f"🎯 Final Answer                 = {p_m3_given_defect:.4f} ({p_m3_given_defect*100:.2f}%)")
```

## 📊 Median & Mode

### Median (درمیانی عدد):
- Data ko pehle sort karein.
- Odd numbers ke liye: $\frac{n+1}{2}$
- Even numbers ke liye: Center ke do numbers ka average
- **Faida:** Outlier resistant (bade number se effect nahi hota).

### Mode (سب سے زیادہ آنے والی value):
- Jo value sabse zyada repeat ho.

```python
import numpy as np
import pandas as pd
from scipy import stats

# Sample marks (with repeat values)
marks = [70, 75, 80, 80, 85, 90]

# --- 1. MEDIAN ---
print(f"Sorted Data : {sorted(marks)}")
median_numpy = np.median(marks)
median_pandas = pd.Series(marks).median()
print(f"📊 Median (NumPy)  : {median_numpy}")
print(f"📊 Median (Pandas) : {median_pandas}")

# --- 2. MODE ---
mode_scipy = stats.mode(marks)
mode_pandas = pd.Series(marks).mode()[0]
print(f"🎯 Mode (SciPy)   : {mode_scipy.mode} (Repeated {mode_scipy.count} times)")
print(f"🎯 Mode (Pandas)  : {mode_pandas}")
```

```python
# ==============================================================================
# 📈 PROBABILITY DISTRIBUTIONS: NORMAL, BINOMIAL & UNIFORM
# ==============================================================================
# 
# ------------------------------------------------------------------------------
# 📚 UPER NOTES (Quick Summary, Rules & Formulas):
# ------------------------------------------------------------------------------
# 1️⃣ NORMAL DISTRIBUTION (Continuous - Bell Curve / Gaussian):
#    - Concept: Zyadatar natural data center (Mean) par jama hota hai aur dono taraf
#      barabar symmetrically ghirta hai. Iska shape ghanti (Bell) jaisa hota hai.
#    - Key Property: Mean = Median = Mode (Symmetric peak at center).
#    - 68-95-99.7 Empirical Rule:
#      * 68.27% data Mean ± 1σ (Std Dev) ke darmiyan hota hai.
#      * 95.45% data Mean ± 2σ ke darmiyan hota hai.
#      * 99.73% data Mean ± 3σ ke darmiyan hota hai.
#    - PDF Formula: f(x) = [1 / (σ * √(2π))] * exp(-0.5 * ((x - μ) / σ)²)
#    - Parameters: μ (Mean/Location), σ (Standard Deviation/Spread).
#    - Examples: Human heights & weights, IQ/exam scores, ML error residuals.
# 
# 2️⃣ BINOMIAL DISTRIBUTION (Discrete - Success / Failure):
#    - Concept: n fixed independent trials mein se k successes aane ka probability model.
#    - 4 Conditions (Bernoulli Trials):
#      1. Har trial ke sirf 2 outcomes: Success (p) ya Failure (1 - p).
#      2. Trials ki tadaad (n) fixed ho.
#      3. Har trial independent ho.
#      4. Success probability (p) constant rahe.
#    - PMF Formula: P(X = k) = nCk * p^k * (1 - p)^(n - k)
#    - Mean: μ = n * p
#    - Variance: σ² = n * p * (1 - p)
#    - Examples: 10 coin tosses mein Heads count karna, MCQ test guessing, quality checks.
# 
# 3️⃣ UNIFORM DISTRIBUTION (Equal Likelihood / Flat Rectangle):
#    - Concept: Kisi given range [a, b] mein har outcome ka chance bilkul barabar hota hai.
#    - Continuous PDF: f(x) = 1 / (b - a)  for a <= x <= b
#    - Mean: μ = (a + b) / 2
#    - Variance: σ² = (b - a)² / 12
#    - Examples: Fair 6-sided dice roll (P=1/6), random number generator.
# ==============================================================================

# ------------------------------------------------------------------------------
# 💻 NICHY CODE (Python Implementation & Visualizations):
# ------------------------------------------------------------------------------
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)  # Reproducible results

# Subplots create karein (1 Row, 3 Columns for side-by-side comparison)
fig, axes = plt.subplots(1, 3, figsize=(18, 5.5))

# ==========================================================
# 1️⃣ NORMAL DISTRIBUTION (Exam Marks: μ=70, σ=10)
# ==========================================================
mu, sigma = 70, 10
normal_data = np.random.normal(loc=mu, scale=sigma, size=10000)

# Histogram (Density = True)
axes[0].hist(normal_data, bins=35, density=True, color="#9b59b6", alpha=0.65, edgecolor="black", label="Simulated Data")

# Theoretical Bell Curve line (PDF)
x_norm = np.linspace(mu - 4*sigma, mu + 4*sigma, 200)
axes[0].plot(x_norm, stats.norm.pdf(x_norm, mu, sigma), color="red", linewidth=2.5, label="Bell Curve (PDF)")

# Mean Line & 68% Rule Shading (μ ± 1σ)
axes[0].axvline(mu, color="black", linestyle="--", linewidth=2, label=f"Mean (μ={mu})")
axes[0].axvspan(mu - sigma, mu + sigma, color="gold", alpha=0.25, label="68% Data (μ ± 1σ)")

axes[0].set_title(f"1️⃣ Normal Distribution\n(μ={mu}, σ={sigma} - Bell Curve)", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Values / Marks", fontsize=10)
axes[0].set_ylabel("Probability Density", fontsize=10)
axes[0].legend(loc="upper right", fontsize=8.5)
axes[0].grid(True, linestyle="--", alpha=0.4)


# ==========================================================
# 2️⃣ BINOMIAL DISTRIBUTION (10 Coin Flips, p=0.5, 10,000 Experiments)
# ==========================================================
n_trials, p_success = 10, 0.5
binomial_data = np.random.binomial(n=n_trials, p=p_success, size=10000)

# Success count probabilities
k_values = np.arange(n_trials + 1)
counts = np.bincount(binomial_data, minlength=n_trials + 1)
probabilities = counts / len(binomial_data)

# Bar chart for discrete trials
axes[1].bar(k_values, probabilities, color="#2ecc71", alpha=0.75, edgecolor="black", label="Simulation (10k Trials)")

# Theoretical PMF (Formula points & line)
theoretical_pmf = stats.binom.pmf(k_values, n_trials, p_success)
axes[1].plot(k_values, theoretical_pmf, "ro-", linewidth=2, markersize=6, label="Formula (PMF)")

axes[1].set_title(f"2️⃣ Binomial Distribution\n(n={n_trials} trials, p={p_success} coin toss)", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Number of Successes / Heads (k)", fontsize=10)
axes[1].set_ylabel("Probability", fontsize=10)
axes[1].set_xticks(k_values)
axes[1].legend(loc="upper right", fontsize=8.5)
axes[1].grid(True, linestyle="--", alpha=0.4)


# ==========================================================
# 3️⃣ UNIFORM DISTRIBUTION (Continuous between a=1 and b=10)
# ==========================================================
a, b = 1, 10
uniform_data = np.random.uniform(a, b, size=10000)

# Histogram (Density = True)
axes[2].hist(uniform_data, bins=30, density=True, color="#3498db", alpha=0.7, edgecolor="black", label="Simulated Data")

# Theoretical Flat PDF Line: 1 / (b - a)
theoretical_pdf_height = 1 / (b - a)
axes[2].axhline(theoretical_pdf_height, color="red", linewidth=2.5, linestyle="--", 
               label=f"Flat PDF = 1/({b}-{a}) = {theoretical_pdf_height:.2f}")

axes[2].set_title(f"3️⃣ Uniform Distribution\n(Equal chance between {a} and {b})", fontsize=12, fontweight="bold")
axes[2].set_xlabel("Values", fontsize=10)
axes[2].set_ylabel("Probability Density", fontsize=10)
axes[2].set_ylim(0, 0.22)
axes[2].legend(loc="upper right", fontsize=8.5)
axes[2].grid(True, linestyle="--", alpha=0.4)


# Layout & Plot Display
plt.tight_layout()
plt.show()


# ==========================================================
# 📊 SUMMARY COMPARISON & VERIFICATION
# ==========================================================
print("=" * 65)
print("📊 PROBABILITY DISTRIBUTIONS - THEORETICAL vs SIMULATED STATS")
print("=" * 65)
print(f"1️⃣ NORMAL   (μ={mu}, σ={sigma}):")
print(f"   - Mean     -> Theoretical: {mu:.2f} | Simulated: {np.mean(normal_data):.2f}")
print(f"   - Variance -> Theoretical: {sigma**2:.2f} | Simulated: {np.var(normal_data):.2f}")
print("-" * 65)
print(f"2️⃣ BINOMIAL (n={n_trials}, p={p_success}):")
print(f"   - Mean     -> Theoretical: {n_trials*p_success:.2f} | Simulated: {np.mean(binomial_data):.2f}")
print(f"   - Variance -> Theoretical: {n_trials*p_success*(1-p_success):.2f} | Simulated: {np.var(binomial_data):.2f}")
print("-" * 65)
print(f"3️⃣ UNIFORM  (a={a}, b={b}):")
print(f"   - Mean     -> Theoretical: {(a+b)/2:.2f} | Simulated: {np.mean(uniform_data):.2f}")
print(f"   - Variance -> Theoretical: {((b-a)**2)/12:.2f} | Simulated: {np.var(uniform_data):.2f}")
print("=" * 65)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Data Points (1 se 5)
x = np.array([1, 2, 3, 4, 5])  # Observation Index
y = np.array([1, 2, 3, 4, 5])  # Values

mean = np.mean(y)
variance = np.var(y)
std_dev = np.std(y)

print(f"Data Points : {y}")
print(f"Mean (μ)    : {mean}")
print(f"Variance(σ²): {variance:.2f}")
print(f"Std Dev (σ) : {std_dev:.4f}")

# 2. Line Plot (plt.plot)
plt.figure(figsize=(10, 6))

# Data Line Plot
plt.plot(x, y, marker="o", markersize=9, color="#1f77b4", linewidth=2.5, label="Data Line (y = [1, 2, 3, 4, 5])")

# Mean Horizontal Line
plt.axhline(mean, color="red", linestyle="--", linewidth=2.5, label=f"Mean Line (μ = {mean:.0f})")

# Standard Deviation Band (Mean ± 1 Std)
plt.axhspan(mean - std_dev, mean + std_dev, color="gold", alpha=0.22, 
            label=f"Std Dev Spread (μ ± σ: {mean-std_dev:.2f} se {mean+std_dev:.2f})")

# Deviations (Har point se Mean tak ki vertical line)
for i in range(len(x)):
    d = y[i] - mean
    # Vertical dotted line point se mean tak
    plt.plot([x[i], x[i]], [mean, y[i]], color="purple", linestyle=":", linewidth=2.2)
    # Distance (d) aur squared distance (d²) ka text label
    plt.text(x[i] + 0.08, (y[i] + mean) / 2, f"d = {d:+.0f}\nd² = {d**2:.0f}", 
             fontsize=10, color="purple", fontweight="bold", va="center")

plt.title("📈 Line Plot: Mean, Deviations (d), Variance (d²) & Std Dev (σ)", fontsize=14, fontweight="bold")
plt.xlabel("Observation (x)", fontsize=12)
plt.ylabel("Values (y)", fontsize=12)
plt.xticks(x)
plt.yticks(range(0, 7))
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(loc="upper left", fontsize=10, frameon=True)
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# 📈 PROBABILITY DISTRIBUTIONS: NORMAL, BINOMIAL & UNIFORM
# ==============================================================================
# 
# ------------------------------------------------------------------------------
# 📚 UPER NOTES (Quick Summary, Rules & Formulas):
# ------------------------------------------------------------------------------
# 1️⃣ NORMAL DISTRIBUTION (Continuous - Bell Curve / Gaussian):
#    - Concept: Zyadatar natural data center (Mean) par jama hota hai aur dono taraf
#      barabar symmetrically ghirta hai. Iska shape ghanti (Bell) jaisa hota hai.
#    - Key Property: Mean = Median = Mode (Symmetric peak at center).
#    - 68-95-99.7 Empirical Rule:
#      * 68.27% data Mean ± 1σ (Std Dev) ke darmiyan hota hai.
#      * 95.45% data Mean ± 2σ ke darmiyan hota hai.
#      * 99.73% data Mean ± 3σ ke darmiyan hota hai.
#    - PDF Formula: f(x) = [1 / (σ * √(2π))] * exp(-0.5 * ((x - μ) / σ)²)
#    - Parameters: μ (Mean/Location), σ (Standard Deviation/Spread).
#    - Examples: Human heights & weights, IQ/exam scores, ML error residuals.
# 
# 2️⃣ BINOMIAL DISTRIBUTION (Discrete - Success / Failure):
#    - Concept: n fixed independent trials mein se k successes aane ka probability model.
#    - 4 Conditions (Bernoulli Trials):
#      1. Har trial ke sirf 2 outcomes: Success (p) ya Failure (1 - p).
#      2. Trials ki tadaad (n) fixed ho.
#      3. Har trial independent ho.
#      4. Success probability (p) constant rahe.
#    - PMF Formula: P(X = k) = nCk * p^k * (1 - p)^(n - k)
#    - Mean: μ = n * p
#    - Variance: σ² = n * p * (1 - p)
#    - Examples: 10 coin tosses mein Heads count karna, MCQ test guessing, quality checks.
# 
# 3️⃣ UNIFORM DISTRIBUTION (Equal Likelihood / Flat Rectangle):
#    - Concept: Kisi given range [a, b] mein har outcome ka chance bilkul barabar hota hai.
#    - Continuous PDF: f(x) = 1 / (b - a)  for a <= x <= b
#    - Mean: μ = (a + b) / 2
#    - Variance: σ² = (b - a)² / 12
#    - Examples: Fair 6-sided dice roll (P=1/6), random number generator.
# ==============================================================================

# ------------------------------------------------------------------------------
# 💻 NICHY CODE (Python Implementation & Visualizations):
# ------------------------------------------------------------------------------
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

np.random.seed(42)  # Reproducible results

# Subplots create karein (1 Row, 3 Columns for side-by-side comparison)
fig, axes = plt.subplots(1, 3, figsize=(18, 5.5))

# ==========================================================
# 1️⃣ NORMAL DISTRIBUTION (Exam Marks: μ=70, σ=10)
# ==========================================================
mu, sigma = 70, 10
normal_data = np.random.normal(loc=mu, scale=sigma, size=10000)

# Histogram (Density = True)
axes[0].hist(normal_data, bins=35, density=True, color="#9b59b6", alpha=0.65, edgecolor="black", label="Simulated Data")

# Theoretical Bell Curve line (PDF)
x_norm = np.linspace(mu - 4*sigma, mu + 4*sigma, 200)
axes[0].plot(x_norm, stats.norm.pdf(x_norm, mu, sigma), color="red", linewidth=2.5, label="Bell Curve (PDF)")

# Mean Line & 68% Rule Shading (μ ± 1σ)
axes[0].axvline(mu, color="black", linestyle="--", linewidth=2, label=f"Mean (μ={mu})")
axes[0].axvspan(mu - sigma, mu + sigma, color="gold", alpha=0.25, label="68% Data (μ ± 1σ)")

axes[0].set_title(f"1️⃣ Normal Distribution\n(μ={mu}, σ={sigma} - Bell Curve)", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Values / Marks", fontsize=10)
axes[0].set_ylabel("Probability Density", fontsize=10)
axes[0].legend(loc="upper right", fontsize=8.5)
axes[0].grid(True, linestyle="--", alpha=0.4)


# ==========================================================
# 2️⃣ BINOMIAL DISTRIBUTION (10 Coin Flips, p=0.5, 10,000 Experiments)
# ==========================================================
n_trials, p_success = 10, 0.5
binomial_data = np.random.binomial(n=n_trials, p=p_success, size=10000)

# Success count probabilities
k_values = np.arange(n_trials + 1)
counts = np.bincount(binomial_data, minlength=n_trials + 1)
probabilities = counts / len(binomial_data)

# Bar chart for discrete trials
axes[1].bar(k_values, probabilities, color="#2ecc71", alpha=0.75, edgecolor="black", label="Simulation (10k Trials)")

# Theoretical PMF (Formula points & line)
theoretical_pmf = stats.binom.pmf(k_values, n_trials, p_success)
axes[1].plot(k_values, theoretical_pmf, "ro-", linewidth=2, markersize=6, label="Formula (PMF)")

axes[1].set_title(f"2️⃣ Binomial Distribution\n(n={n_trials} trials, p={p_success} coin toss)", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Number of Successes / Heads (k)", fontsize=10)
axes[1].set_ylabel("Probability", fontsize=10)
axes[1].set_xticks(k_values)
axes[1].legend(loc="upper right", fontsize=8.5)
axes[1].grid(True, linestyle="--", alpha=0.4)


# ==========================================================
# 3️⃣ UNIFORM DISTRIBUTION (Continuous between a=1 and b=10)
# ==========================================================
a, b = 1, 10
uniform_data = np.random.uniform(a, b, size=10000)

# Histogram (Density = True)
axes[2].hist(uniform_data, bins=30, density=True, color="#3498db", alpha=0.7, edgecolor="black", label="Simulated Data")

# Theoretical Flat PDF Line: 1 / (b - a)
theoretical_pdf_height = 1 / (b - a)
axes[2].axhline(theoretical_pdf_height, color="red", linewidth=2.5, linestyle="--", 
               label=f"Flat PDF = 1/({b}-{a}) = {theoretical_pdf_height:.2f}")

axes[2].set_title(f"3️⃣ Uniform Distribution\n(Equal chance between {a} and {b})", fontsize=12, fontweight="bold")
axes[2].set_xlabel("Values", fontsize=10)
axes[2].set_ylabel("Probability Density", fontsize=10)
axes[2].set_ylim(0, 0.22)
axes[2].legend(loc="upper right", fontsize=8.5)
axes[2].grid(True, linestyle="--", alpha=0.4)


# Layout & Plot Display
plt.tight_layout()
plt.show()


# ==========================================================
# 📊 SUMMARY COMPARISON & VERIFICATION
# ==========================================================
print("=" * 65)
print("📊 PROBABILITY DISTRIBUTIONS - THEORETICAL vs SIMULATED STATS")
print("=" * 65)
print(f"1️⃣ NORMAL   (μ={mu}, σ={sigma}):")
print(f"   - Mean     -> Theoretical: {mu:.2f} | Simulated: {np.mean(normal_data):.2f}")
print(f"   - Variance -> Theoretical: {sigma**2:.2f} | Simulated: {np.var(normal_data):.2f}")
print("-" * 65)
print(f"2️⃣ BINOMIAL (n={n_trials}, p={p_success}):")
print(f"   - Mean     -> Theoretical: {n_trials*p_success:.2f} | Simulated: {np.mean(binomial_data):.2f}")
print(f"   - Variance -> Theoretical: {n_trials*p_success*(1-p_success):.2f} | Simulated: {np.var(binomial_data):.2f}")
print("-" * 65)
print(f"3️⃣ UNIFORM  (a={a}, b={b}):")
print(f"   - Mean     -> Theoretical: {(a+b)/2:.2f} | Simulated: {np.mean(uniform_data):.2f}")
print(f"   - Variance -> Theoretical: {((b-a)**2)/12:.2f} | Simulated: {np.var(uniform_data):.2f}")
print("=" * 65)
```
