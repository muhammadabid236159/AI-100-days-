## 1️⃣ Limits ($\lim_{x \to a} f(x)$)

### 📖 Concept (Simple English):
- **Limit** ka matlab hai: *jab $x$ kisi value $a$ ke kareeb kareeb jata hai (lekin exactly $a$ nahi hota) tab $f(x)$ ki value kya hoti hai?*
- Ye Calculus ki buniyad (foundation) hai — derivative aur integral dono limits se bante hain.

### 📐 Formulas:
$$\lim_{x \to a} f(x) = L$$

**Derivative ki limit definition:**
$$f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$$

### 🤖 ML Connection:
- Gradient Descent mein step size $h \to 0$ ki limit se derivative calculate hoti hai.
- Neural network activation functions (ReLU, Sigmoid) ke kinaray pe limits ka concept use hota hai.

```python
# ==========================================================
# 1. LIMITS: NUMERICAL & VISUAL DEMONSTRATION
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt

# Numerically approaching a limit: lim(x->2) of f(x) = x^2 - 4 / (x - 2)
# Note: f(x) is undefined at x=2 (0/0), but limit exists!
def f(x):
    return (x**2 - 4) / (x - 2)  # = (x+2)(x-2)/(x-2) = x+2

print('Numerical Approach to limit as x -> 2:')
print(f'{"x":<12} {"f(x) = (x²-4)/(x-2)":<25}')
print('-' * 38)
for x_val in [1.9, 1.99, 1.999, 1.9999, 2.0001, 2.001, 2.01, 2.1]:
    print(f'{x_val:<12.4f} {f(x_val):<25.6f}')

print('\n✅ As x → 2, f(x) → 4.0  (Limit = 4)')

# Visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))

# Plot 1: The function (x^2-4)/(x-2)
x_left = np.linspace(0, 1.98, 200)
x_right = np.linspace(2.02, 4, 200)

ax1.plot(x_left, f(x_left), color='#1f77b4', linewidth=3, label='f(x) = (x²-4)/(x-2)')
ax1.plot(x_right, f(x_right), color='#1f77b4', linewidth=3)
ax1.scatter([2], [4], color='white', edgecolors='red', s=120, zorder=5, linewidths=2.5,
            label='Hole at x=2 (undefined!)')
ax1.axvline(x=2, color='red', linestyle='--', alpha=0.5, label='x = 2')
ax1.axhline(y=4, color='green', linestyle='--', alpha=0.5, label='Limit = 4')
ax1.annotate('lim f(x) = 4\nx→2', xy=(2, 4), xytext=(2.5, 2.5),
             arrowprops=dict(arrowstyle='->', color='green'), fontsize=12,
             fontweight='bold', color='green')
ax1.set_xlabel('x', fontweight='bold')
ax1.set_ylabel('f(x)', fontweight='bold')
ax1.set_title('Limit: lim(x→2) of (x²-4)/(x-2) = 4', fontsize=12, fontweight='bold')
ax1.legend()
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.set_xlim([0, 4])
ax1.set_ylim([0, 7])

# Plot 2: Derivative Limit Definition h->0
def g(x): return x**2
x0 = 2.0
x_range = np.linspace(0, 4, 300)

ax2.plot(x_range, g(x_range), color='#1f77b4', linewidth=3, label='f(x) = x²')
colors = ['#e74c3c', '#e67e22', '#27ae60']
h_vals = [1.5, 0.75, 0.2]
for h, color in zip(h_vals, colors):
    slope = (g(x0 + h) - g(x0)) / h
    x_line = np.linspace(x0 - 0.5, x0 + h + 0.5, 100)
    y_line = g(x0) + slope * (x_line - x0)
    ax2.plot(x_line, y_line, color=color, linewidth=2, linestyle='--', label=f'Secant h={h} (slope={slope:.1f})')

# True tangent at x=2 (slope = 2*x = 4)
x_tan = np.linspace(0.5, 3.5, 100)
ax2.plot(x_tan, g(x0) + 4*(x_tan - x0), color='black', linewidth=2.5, label='Tangent h→0 (slope=4)')
ax2.scatter([x0], [g(x0)], color='red', s=100, zorder=5)
ax2.set_xlim([0, 4])
ax2.set_ylim([-2, 16])
ax2.set_xlabel('x', fontweight='bold')
ax2.set_ylabel('f(x)', fontweight='bold')
ax2.set_title('Derivative Limit: Secant → Tangent as h→0', fontsize=12, fontweight='bold')
ax2.legend(fontsize=9)
ax2.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

## 2️⃣ Derivatives (Differentiation) — $f'(x)$

### 📖 Concept (Simple English):
- **Derivative** batata hai ke function **kitni tezi se change ho raha hai** — yaani slope ya rate of change.
- Geometrically: derivative $f'(x)$ **tangent line ka slope** hai at any point $x$.
- Agar $f'(x) > 0$: function increase ho raha hai (curve upar ja rahi hai).
- Agar $f'(x) < 0$: function decrease ho raha hai (curve neeche ja rahi hai).
- Agar $f'(x) = 0$: **minima ya maxima** (flat point — Gradient Descent ka goal!).

### 📐 Common Derivative Rules:
| Rule | Formula |
|------|----------|
| **Power Rule** | $\frac{d}{dx} x^n = n x^{n-1}$ |
| **Constant Rule** | $\frac{d}{dx} c = 0$ |
| **Sum Rule** | $\frac{d}{dx}[f + g] = f' + g'$ |
| **Product Rule** | $\frac{d}{dx}[f \cdot g] = f'g + fg'$ |
| **Chain Rule** | $\frac{d}{dx}[f(g(x))] = f'(g(x)) \cdot g'(x)$ |
| **Exponential** | $\frac{d}{dx} e^x = e^x$ |
| **Logarithm** | $\frac{d}{dx} \ln(x) = \frac{1}{x}$ |

### 🤖 ML Connection:
- **Loss function ka derivative** = gradient = direction of steepest ascent/descent.
- **Backpropagation** = chain rule applied layer by layer in a neural network.
- Finding **minimum of Loss** = finding where derivative/gradient = 0.

```python
# ==========================================================
# 2. DERIVATIVES: SYMBOLIC & NUMERICAL + VISUALIZATION
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

x = sp.Symbol('x')

# Symbolic Derivatives using SymPy
functions = [
    ('f(x) = x³',         x**3,         '3x²'),
    ('f(x) = x² + 3x',    x**2 + 3*x,   '2x + 3'),
    ('f(x) = sin(x)',      sp.sin(x),    'cos(x)'),
    ('f(x) = e^x',        sp.exp(x),    'e^x'),
    ('f(x) = ln(x)',       sp.ln(x),     '1/x'),
    ('f(x) = x²·e^x',     x**2 * sp.exp(x), '2x·eˣ + x²·eˣ (Product Rule)'),
]

print(f'{"Function":<20} {"Derivative f\'(x)":<35} {"Expected"}')
print('-' * 75)
for name, func, expected in functions:
    deriv = sp.diff(func, x)
    print(f'{name:<20} {str(sp.simplify(deriv)):<35} ✅ {expected}')

# Visualization
fig, axes = plt.subplots(1, 3, figsize=(17, 5.5))

# Plot 1: f(x) = x³ and f'(x) = 3x²
x_arr = np.linspace(-2.5, 2.5, 300)
axes[0].plot(x_arr, x_arr**3, color='#1f77b4', linewidth=3, label='f(x) = x³')
axes[0].plot(x_arr, 3*x_arr**2, color='#d62728', linewidth=3, linestyle='--', label="f'(x) = 3x²")
axes[0].axhline(0, color='black', linewidth=0.8)
axes[0].axvline(0, color='black', linewidth=0.8)
axes[0].fill_between(x_arr, 3*x_arr**2, alpha=0.1, color='#d62728')
axes[0].set_title('Power Rule: d/dx(x³) = 3x²', fontweight='bold')
axes[0].legend()
axes[0].grid(True, linestyle=':', alpha=0.6)
axes[0].set_ylim([-10, 10])

# Plot 2: f(x) = sin(x) and f'(x) = cos(x)
x_arr2 = np.linspace(-2*np.pi, 2*np.pi, 400)
axes[1].plot(x_arr2, np.sin(x_arr2), color='#2ca02c', linewidth=3, label='f(x) = sin(x)')
axes[1].plot(x_arr2, np.cos(x_arr2), color='#ff7f0e', linewidth=3, linestyle='--', label="f'(x) = cos(x)")
axes[1].axhline(0, color='black', linewidth=0.8)
axes[1].set_title('Trig: d/dx(sin x) = cos x', fontweight='bold')
axes[1].legend()
axes[1].grid(True, linestyle=':', alpha=0.6)

# Plot 3: Tangent line at different points of f(x) = x²
x_arr3 = np.linspace(-3, 3, 300)
axes[2].plot(x_arr3, x_arr3**2, color='#9467bd', linewidth=3, label='f(x) = x²')
tangent_points = [(-2, 'red'), (0, 'green'), (1.5, 'orange')]
for x0, color in tangent_points:
    slope = 2 * x0
    x_tan = np.linspace(x0 - 1, x0 + 1, 50)
    y_tan = x0**2 + slope * (x_tan - x0)
    axes[2].plot(x_tan, y_tan, color=color, linewidth=2.5, linestyle='--', label=f'Tangent at x={x0} (slope={slope})')
    axes[2].scatter([x0], [x0**2], color=color, s=80, zorder=5)
axes[2].set_ylim([-1, 9])
axes[2].set_title('Tangent Lines: Slope = f\'(x) = 2x', fontweight='bold')
axes[2].legend(fontsize=9)
axes[2].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

## 3️⃣ Partial Derivatives ($\frac{\partial f}{\partial x}$, $\frac{\partial f}{\partial y}$)

### 📖 Concept (Simple English):
- Jab function mein **ek se zyada variables** hon ($x, y, z, \dots$), to har variable ke respect mein alag alag derivative nikalte hain.
- **Partial Derivative** mein: **sirf ek variable change hota hai**, baaki sab ko **constant** maan lete hain.

### 📐 Formulas:
$$f(x, y) = x^2 + 3xy + y^3$$
$$\frac{\partial f}{\partial x} = 2x + 3y \quad \text{(y ko constant maana)}$$
$$\frac{\partial f}{\partial y} = 3x + 3y^2 \quad \text{(x ko constant maana)}$$

### 🤖 ML Connection:
- Neural network mein **thousands of weights** ($w_1, w_2, \dots, w_n$) hote hain — Loss function ka har weight ke respect mein partial derivative nikalna backpropagation hai!
- $\frac{\partial L}{\partial w_j}$ = weight $w_j$ ko ek unit change karne se Loss mein kitna change aata hai.

```python
# ==========================================================
# 3. PARTIAL DERIVATIVES: SYMBOLIC & 3D VISUALIZATION
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import sympy as sp

x, y = sp.symbols('x y')

# Function: f(x, y) = x^2 + y^2 (simple bowl)
f_sym = x**2 + y**2
df_dx = sp.diff(f_sym, x)
df_dy = sp.diff(f_sym, y)

print('Function: f(x, y) = x² + y²')
print(f'∂f/∂x = {df_dx}  (slope in X direction)')
print(f'∂f/∂y = {df_dy}  (slope in Y direction)')

# Another example
f2 = x**2 * y + sp.sin(x) + y**3
print('\nFunction: f(x, y) = x²y + sin(x) + y³')
print(f'∂f/∂x = {sp.diff(f2, x)}')
print(f'∂f/∂y = {sp.diff(f2, y)}')

# 3D Visualization
fig = plt.figure(figsize=(16, 6))

x_arr = np.linspace(-3, 3, 60)
y_arr = np.linspace(-3, 3, 60)
X, Y = np.meshgrid(x_arr, y_arr)
Z = X**2 + Y**2  # f(x,y)
dZ_dx = 2 * X    # ∂f/∂x
dZ_dy = 2 * Y    # ∂f/∂y

# Plot 1: 3D Surface f(x,y) = x² + y²
ax1 = fig.add_subplot(131, projection='3d')
surf = ax1.plot_surface(X, Y, Z, cmap='viridis', alpha=0.85)
ax1.set_xlabel('X', fontweight='bold')
ax1.set_ylabel('Y', fontweight='bold')
ax1.set_zlabel('f(x,y)', fontweight='bold')
ax1.set_title('f(x,y) = x² + y²\n(Bowl Surface)', fontweight='bold')
plt.colorbar(surf, ax=ax1, shrink=0.5)

# Plot 2: ∂f/∂x = 2x (slope in X direction)
ax2 = fig.add_subplot(132, projection='3d')
surf2 = ax2.plot_surface(X, Y, dZ_dx, cmap='RdBu', alpha=0.85)
ax2.set_xlabel('X', fontweight='bold')
ax2.set_ylabel('Y', fontweight='bold')
ax2.set_zlabel('∂f/∂x', fontweight='bold')
ax2.set_title('∂f/∂x = 2x\n(X-direction Slope)', fontweight='bold')
plt.colorbar(surf2, ax=ax2, shrink=0.5)

# Plot 3: Gradient field (arrows)
ax3 = fig.add_subplot(133)
skip = 5
ax3.contourf(X, Y, Z, levels=20, cmap='viridis', alpha=0.7)
ax3.quiver(X[::skip, ::skip], Y[::skip, ::skip],
           dZ_dx[::skip, ::skip], dZ_dy[::skip, ::skip],
           color='white', scale=60, alpha=0.9)
ax3.set_xlabel('X', fontweight='bold')
ax3.set_ylabel('Y', fontweight='bold')
ax3.set_title('Gradient Field ∇f = (∂f/∂x, ∂f/∂y)\nArrows point uphill!', fontweight='bold')

plt.tight_layout()
plt.show()
```

## 4️⃣ Gradient ($\nabla f$) — Direction of Steepest Ascent

### 📖 Concept (Simple English):
- **Gradient** ek **vector** hai jo tamam partial derivatives ko ek jagah collect karta hai:
  $$\nabla f = \left(\frac{\partial f}{\partial x_1},\ \frac{\partial f}{\partial x_2},\ \dots,\ \frac{\partial f}{\partial x_n}\right)$$
- Gradient hamesha **steepest ascent (sab se tezi se upar) ki direction** batata hai.
- **Negative Gradient** ($-\nabla f$) = steepest descent (sab se tezi se neeche) ki direction.

### 📐 Example:
$$f(x, y) = x^2 + y^2$$
$$\nabla f = \begin{bmatrix} 2x \\ 2y \end{bmatrix}$$

**At point $(1, 2)$:**
$$\nabla f(1, 2) = \begin{bmatrix} 2 \\ 4 \end{bmatrix}$$
Matlab: is point se $x$ direction mein $2$ aur $y$ direction mein $4$ ki speed se function badhta hai.

### 🤖 ML Connection:
- Neural network ka **Loss function** $(w_1, w_2, \dots, w_n)$ weights ka function hai.
- Gradient $\nabla L$ batata hai ke kaunsi direction mein weights change karne se Loss badhta hai.
- Hum **negative gradient** direction mein move karte hain (**Gradient Descent**) taake Loss kam ho!

```python
# ==========================================================
# 4. GRADIENT: COMPUTATION & VECTOR FIELD VISUALIZATION
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt

# f(x, y) = x^2 + y^2 --- gradient = [2x, 2y]
def f(x, y):   return x**2 + y**2
def grad_x(x): return 2 * x
def grad_y(y): return 2 * y

# Manual Gradient at a point
points = [(1, 2), (-2, 1), (3, -1), (0, 0)]
print('Gradient of f(x,y) = x² + y²:')
print(f'{"Point (x,y)":<15} {"f(x,y)":<12} {"∇f = (∂f/∂x, ∂f/∂y)":<30} {"Magnitude |∇f|"}')
print('-' * 70)
for px, py in points:
    gx = grad_x(px)
    gy = grad_y(py)
    mag = np.sqrt(gx**2 + gy**2)
    print(f'({px}, {py})       {f(px,py):<12} ({gx}, {gy})            {mag:.4f}')

print('\n✅ At (0,0): gradient = (0,0) → Global Minimum!')

# Visualization: Gradient field on contour map
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

x_arr = np.linspace(-3, 3, 50)
y_arr = np.linspace(-3, 3, 50)
X, Y = np.meshgrid(x_arr, y_arr)
Z = X**2 + Y**2
GX = 2 * X
GY = 2 * Y

# Plot 1: Gradient (Ascent) Field
cp1 = axes[0].contourf(X, Y, Z, levels=15, cmap='YlOrRd', alpha=0.75)
plt.colorbar(cp1, ax=axes[0], label='f(x,y)')
skip = 4
axes[0].quiver(X[::skip, ::skip], Y[::skip, ::skip],
               GX[::skip, ::skip], GY[::skip, ::skip],
               color='darkred', scale=50, alpha=0.85)
axes[0].scatter([0], [0], color='white', s=200, zorder=5, label='Minimum (0,0)')
axes[0].set_title('∇f = Gradient (Points UPHILL)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('X', fontweight='bold')
axes[0].set_ylabel('Y', fontweight='bold')
axes[0].legend()

# Plot 2: Negative Gradient (Descent) Field - what gradient descent follows!
cp2 = axes[1].contourf(X, Y, Z, levels=15, cmap='YlGn', alpha=0.75)
plt.colorbar(cp2, ax=axes[1], label='f(x,y)')
axes[1].quiver(X[::skip, ::skip], Y[::skip, ::skip],
               -GX[::skip, ::skip], -GY[::skip, ::skip],
               color='darkgreen', scale=50, alpha=0.85)
axes[1].scatter([0], [0], color='red', s=200, zorder=5, label='Minimum (0,0) ← Target')
axes[1].set_title('-∇f = Negative Gradient\n(Gradient Descent Direction!)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('X', fontweight='bold')
axes[1].set_ylabel('Y', fontweight='bold')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 5️⃣ Gradient Descent — Heart of Neural Network Training

### 📖 Concept (Simple English):
- **Gradient Descent** aik iterative optimization algorithm hai jo **Loss function ko minimize** karta hai.
- **Analogy:** Sochiye aap andhere pahar pe hain aur neeche utarna hai — har qadam negative gradient direction mein uthate ho (sab se tezi se neeche ki taraf).

### 📐 Update Rule:
$$\theta_{\text{new}} = \theta_{\text{old}} - \alpha \cdot \nabla_{\theta} L(\theta)$$

Where:
- $\theta$ = model weights (parameters)
- $\alpha$ = **Learning Rate** (step size per iteration)
- $\nabla_{\theta} L$ = gradient of Loss with respect to weights

### ⚙️ Types of Gradient Descent:
| Type | Description |
|------|-------------|
| **Batch GD** | Puri dataset ka gradient use karta hai (accurate lekin slow) |
| **Stochastic GD (SGD)** | Ek ek sample ka gradient use karta hai (fast lekin noisy) |
| **Mini-batch GD** | Small batches ka gradient (balance between both — most common!) |

### 🤖 ML Connection:
- PyTorch aur TensorFlow mein `optimizer.step()` Gradient Descent hi karta hai!
- **Adam, RMSprop, Momentum** sab Gradient Descent ke advanced variants hain.

```python
# ==========================================================
# 5. GRADIENT DESCENT: STEP-BY-STEP SIMULATION
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt

# Loss function: L(w) = w^2 + 2w + 1 = (w+1)^2 → Minimum at w = -1
def loss(w):       return w**2 + 2*w + 1
def grad_loss(w):  return 2*w + 2        # dL/dw

# Gradient Descent parameters
learning_rates = [0.1, 0.5, 0.9]
w_start = 4.0
n_steps = 25

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

w_range = np.linspace(-3, 5, 300)

for ax, lr in zip(axes, learning_rates):
    w = w_start
    history_w = [w]
    history_loss = [loss(w)]

    for _ in range(n_steps):
        g = grad_loss(w)
        w = w - lr * g    # Update rule!
        history_w.append(w)
        history_loss.append(loss(w))

    # Plot loss curve
    ax.plot(w_range, loss(w_range), color='#1f77b4', linewidth=3, label='L(w) = (w+1)²')

    # Plot descent steps
    for i in range(len(history_w) - 1):
        ax.annotate('', xy=(history_w[i+1], loss(history_w[i+1])),
                    xytext=(history_w[i], loss(history_w[i])),
                    arrowprops=dict(arrowstyle='->', color='red', lw=1.5))

    ax.scatter(history_w, [loss(w) for w in history_w],
               color='red', s=50, zorder=5, alpha=0.8)
    ax.scatter([-1], [0], color='green', s=200, zorder=6, marker='*', label='Minimum w=-1')

    ax.set_xlabel('Weight w', fontweight='bold')
    ax.set_ylabel('Loss L(w)', fontweight='bold')
    ax.set_title(f'Gradient Descent (lr={lr})\nFinal w={history_w[-1]:.4f}, Loss={history_loss[-1]:.6f}',
                 fontweight='bold')
    ax.legend()
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.set_xlim([-3, 5])
    ax.set_ylim([-0.5, 25])

plt.tight_layout()
plt.show()

# Print step-by-step for lr=0.1
print('\nStep-by-step Gradient Descent (lr=0.1):')
print(f'{"Step":<6} {"w":<12} {"Gradient":<15} {"Loss"}')
print('-' * 45)
w = w_start
for step in range(10):
    g = grad_loss(w)
    print(f'{step:<6} {w:<12.6f} {g:<15.6f} {loss(w):.6f}')
    w = w - 0.1 * g
print(f'...\nFinal w ≈ {w:.6f} (True minimum = -1.0) ✅')
```

## 6️⃣ Integration — Area Under the Curve ($\int f(x) dx$)

### 📖 Concept (Simple English):
- **Integration** differentiation ka ulta (inverse) operation hai — **anti-derivative**.
- Geometrically: definite integral = function aur x-axis ke darmiyan **area**.
- **Fundamental Theorem of Calculus:**
  $$\int_a^b f(x) dx = F(b) - F(a) \quad \text{where } F'(x) = f(x)$$

### 📐 Common Integrals:
| Integral | Result |
|----------|--------|
| $\int x^n dx$ | $\frac{x^{n+1}}{n+1} + C$ |
| $\int e^x dx$ | $e^x + C$ |
| $\int \frac{1}{x} dx$ | $\ln|x| + C$ |
| $\int \cos(x) dx$ | $\sin(x) + C$ |

### 🤖 ML Connection:
- **Probability & Statistics:** Probability distributions mein $\int_{-\infty}^{\infty} p(x) dx = 1$.
- **Expected Value:** $E[X] = \int x \cdot p(x) dx$ continuous distributions ke liye.
- **KL Divergence** (used in GANs, VAEs): $D_{KL}(P \| Q) = \int P(x) \log \frac{P(x)}{Q(x)} dx$.

```python
# ==========================================================
# 6. INTEGRATION: SYMBOLIC + NUMERICAL + VISUALIZATION
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from scipy import integrate

x = sp.Symbol('x')

# Symbolic Integration
integrals = [
    ('x³',        x**3,      'x⁴/4 + C'),
    ('sin(x)',     sp.sin(x), '-cos(x) + C'),
    ('eˣ',        sp.exp(x), 'eˣ + C'),
    ('1/x',       1/x,       'ln|x| + C'),
    ('2x + 3',    2*x + 3,   'x² + 3x + C'),
]

print('Indefinite Integrals:')
print(f'{"f(x)":<15} {"∫f(x)dx":<35} Expected')
print('-' * 65)
for name, func, expected in integrals:
    result = sp.integrate(func, x)
    print(f'{name:<15} {str(result):<35} ✅ {expected}')

# Definite Integral numerically
def f(x): return x**2
result_numeric, err = integrate.quad(f, 0, 3)
result_exact = 3**3 / 3  # = 9
print(f'\n∫₀³ x² dx = {result_numeric:.6f} (Exact = {result_exact})  ✅')

# Visualization: Area under curve
fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))

# Plot 1: Area under f(x) = x²
x_arr = np.linspace(0, 4, 300)
x_fill = np.linspace(0, 3, 300)
axes[0].plot(x_arr, x_arr**2, color='#1f77b4', linewidth=3, label='f(x) = x²')
axes[0].fill_between(x_fill, x_fill**2, alpha=0.35, color='#1f77b4', label='Area = ∫₀³ x² dx = 9')
axes[0].axvline(x=0, color='green', linestyle='--', alpha=0.7)
axes[0].axvline(x=3, color='red', linestyle='--', alpha=0.7)
axes[0].text(1.0, 3.5, '∫₀³ x² dx = 9', fontsize=14, fontweight='bold', color='#1f77b4')
axes[0].set_xlabel('x', fontweight='bold')
axes[0].set_ylabel('f(x)', fontweight='bold')
axes[0].set_title('Definite Integral as Area Under Curve', fontweight='bold')
axes[0].legend()
axes[0].grid(True, linestyle=':', alpha=0.6)

# Plot 2: Riemann Sum Approximation (how computers numerically integrate)
n_rects = 10
x_arr2 = np.linspace(0, 3, 300)
axes[1].plot(x_arr2, x_arr2**2, color='#1f77b4', linewidth=3, label='f(x) = x²')
dx = 3.0 / n_rects
riemann_sum = 0
for i in range(n_rects):
    xi = i * dx
    height = xi**2
    riemann_sum += height * dx
    rect = plt.Rectangle((xi, 0), dx, height, alpha=0.4, facecolor='orange', edgecolor='darkorange')
    axes[1].add_patch(rect)
axes[1].set_xlabel('x', fontweight='bold')
axes[1].set_ylabel('f(x)', fontweight='bold')
axes[1].set_title(f'Riemann Sum Approx (n={n_rects} rectangles)\nApprox Area ≈ {riemann_sum:.4f} (Exact = 9)', fontweight='bold')
axes[1].legend()
axes[1].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

## 7️⃣ Chain Rule & Backpropagation — How Neural Networks Learn

### 📖 Concept (Simple English):
- **Chain Rule:** Jab functions ek doosre ke andar ho (composite functions), derivative nikalne ke liye Chain Rule use hoti hai.
  $$\frac{d}{dx} f(g(x)) = f'(g(x)) \cdot g'(x)$$
- **Backpropagation = Chain Rule applied from output layer back to input layer!**

### 📐 Example — Simple Neural Network:
Forward Pass:
$$z = wx + b \quad \to \quad a = \sigma(z) \quad \to \quad L = (a - y)^2$$

Backward Pass (using Chain Rule):
$$\frac{\partial L}{\partial w} = \frac{\partial L}{\partial a} \cdot \frac{\partial a}{\partial z} \cdot \frac{\partial z}{\partial w}$$

Where:
- $\frac{\partial L}{\partial a} = 2(a - y)$ (Loss ke sath)
- $\frac{\partial a}{\partial z} = \sigma'(z) = \sigma(z)(1 - \sigma(z))$ (Sigmoid derivative)
- $\frac{\partial z}{\partial w} = x$ (Linear layer derivative)

### 🤖 ML Connection:
- PyTorch ka `loss.backward()` = Chain Rule automatically apply karta hai har layer mein!
- Deeper networks mein **Vanishing Gradient Problem**: Chain Rule se gradients multiply hote hain — agar sab < 1 hon to bahut chote ho jaate hain — isliye **ReLU activation** aur **Batch Normalization** use karte hain.

```python
# ==========================================================
# 7. CHAIN RULE & BACKPROPAGATION MANUAL SIMULATION
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt

# Sigmoid activation and its derivative
def sigmoid(z):        return 1 / (1 + np.exp(-z))
def sigmoid_deriv(z):  return sigmoid(z) * (1 - sigmoid(z))

# Manual Backprop for a single neuron: L = (sigma(wx + b) - y)^2
x_val = 2.0
y_val = 1.0  # True label
w = 0.5      # Initial weight
b = 0.1      # Initial bias
lr = 0.3

print('=== Manual Forward Pass + Backpropagation ===')
history_loss = []
history_w = []
history_b = []

for step in range(20):
    # 1. FORWARD PASS
    z = w * x_val + b               # Linear combo
    a = sigmoid(z)                   # Activation
    L = (a - y_val)**2              # Loss (MSE)

    history_loss.append(L)
    history_w.append(w)
    history_b.append(b)

    # 2. BACKWARD PASS (Chain Rule)
    dL_da = 2 * (a - y_val)         # ∂L/∂a
    da_dz = sigmoid_deriv(z)        # ∂a/∂z  (sigmoid derivative)
    dz_dw = x_val                   # ∂z/∂w = x
    dz_db = 1.0                     # ∂z/∂b = 1

    dL_dw = dL_da * da_dz * dz_dw  # Chain Rule!
    dL_db = dL_da * da_dz * dz_db

    # 3. WEIGHT UPDATE
    w = w - lr * dL_dw
    b = b - lr * dL_db

    if step < 5 or step == 19:
        print(f'Step {step:2d}: z={z:.4f}, a={a:.4f}, L={L:.6f}, ∂L/∂w={dL_dw:.6f}, w={w:.6f}')

print(f'\nFinal w={w:.6f}, b={b:.6f}, Loss={history_loss[-1]:.8f}')
print('✅ Loss converged to near 0!')

# Visualization
fig, axes = plt.subplots(1, 3, figsize=(17, 5))

# Plot 1: Loss over iterations
axes[0].plot(history_loss, color='#d62728', linewidth=3, marker='o', markersize=4)
axes[0].set_xlabel('Iteration', fontweight='bold')
axes[0].set_ylabel('Loss L', fontweight='bold')
axes[0].set_title('Loss Decreasing via Backpropagation', fontweight='bold')
axes[0].grid(True, linestyle=':', alpha=0.6)
axes[0].fill_between(range(len(history_loss)), history_loss, alpha=0.2, color='red')

# Plot 2: Weight over iterations
axes[1].plot(history_w, color='#1f77b4', linewidth=3, marker='s', markersize=4)
axes[1].set_xlabel('Iteration', fontweight='bold')
axes[1].set_ylabel('Weight w', fontweight='bold')
axes[1].set_title('Weight w Converging', fontweight='bold')
axes[1].grid(True, linestyle=':', alpha=0.6)

# Plot 3: Sigmoid and its derivative
z_arr = np.linspace(-6, 6, 300)
axes[2].plot(z_arr, sigmoid(z_arr), color='#2ca02c', linewidth=3, label='σ(z) = 1/(1+e⁻ᶻ)')
axes[2].plot(z_arr, sigmoid_deriv(z_arr), color='#ff7f0e', linewidth=3, linestyle='--', label="σ'(z) = σ(z)·(1-σ(z))")
axes[2].axhline(0.25, color='gray', linestyle=':', alpha=0.6)
axes[2].text(0.1, 0.27, 'Max σ\'(z) = 0.25', fontsize=11, color='gray')
axes[2].set_xlabel('z', fontweight='bold')
axes[2].set_ylabel('Value', fontweight='bold')
axes[2].set_title('Sigmoid & Its Derivative\n(Used in Chain Rule Backprop)', fontweight='bold')
axes[2].legend()
axes[2].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
