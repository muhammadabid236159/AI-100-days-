```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Seaborn theme set karo
sns.set_theme(style="whitegrid")
```

```python
sns.get_dataset_names()
```

```python
flights = sns.load_dataset("flights")
flights
```

```python

sns.barplot(
    data=flights,
    x='month',
    y="passengers",   
)
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
sns.lineplot(x=x, y=y)
```

```python
tips=sns.load_dataset("tips")
sns.barplot(
    data=tips,
    x='day',
    y='tip',
    hue='sex'
)
```

```python
sns.boxplot(
    data=tips,
    x='day',
    y='tip',
    hue='sex'
)
```
