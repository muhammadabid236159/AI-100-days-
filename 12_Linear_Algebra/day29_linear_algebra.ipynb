## 1️⃣ Distance Between Two Points

### 📖 Concept (Simple English):
The distance between two points in a 2D plane is the shortest straight-line length connecting them. It is calculated using the Pythagorean Theorem.

- **Point 1:** $P_1 = (x_1, y_1)$
- **Point 2:** $P_2 = (x_2, y_2)$

### 📐 Formula (Euclidean Distance / $L_2$ Norm):
$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

### 🤖 Machine Learning Connection:
Euclidean distance is the fundamental metric used in **K-Nearest Neighbors (KNN)** classification, **K-Means Clustering**, and error loss calculations.

```python
# ==========================================================
# 1. DISTANCE BETWEEN TWO POINTS & VISUALIZATION
# ==========================================================

# Define two points
p1 = np.array([2, 3])
p2 = np.array([8, 11])

# Calculate Euclidean Distance using formula & NumPy
distance_formula = np.sqrt((p2[0] - p1[0])**2 + (p2[1] - p1[1])**2)
distance_numpy = np.linalg.norm(p2 - p1)

print(f"Point 1 (x1, y1)   : {tuple(p1)}")
print(f"Point 2 (x2, y2)   : {tuple(p2)}")
print(f"Distance (Formula) : {distance_formula:.4f}")
print(f"Distance (np.norm) : {distance_numpy:.4f}")

# --- Visualization ---
plt.figure(figsize=(8, 6))

# Plot the points
plt.scatter([p1[0], p2[0]], [p1[1], p2[1]], color="red", s=120, zorder=5, label="Points P1 & P2")

# Plot the direct distance line
plt.plot([p1[0], p2[0]], [p1[1], p2[1]], color="#2b5c8f", linewidth=2.5, linestyle="-", label=f"Distance d = {distance_numpy:.2f}")

# Plot right-angled triangle components (dx, dy)
plt.plot([p1[0], p2[0]], [p1[1], p1[1]], color="gray", linestyle="--", alpha=0.7, label=f"Δx = {p2[0] - p1[0]}")
plt.plot([p2[0], p2[0]], [p1[1], p2[1]], color="gray", linestyle=":", alpha=0.7, label=f"Δy = {p2[1] - p1[1]}")

# Annotations
plt.text(p1[0] - 0.4, p1[1] + 0.4, f"P1 (2, 3)", fontsize=11, fontweight="bold")
plt.text(p2[0] + 0.2, p2[1], f"P2 (8, 11)", fontsize=11, fontweight="bold")
plt.text((p1[0] + p2[0])/2 - 0.5, (p1[1] + p2[1])/2 + 0.6, f"d = {distance_numpy:.2f}", fontsize=11, color="#2b5c8f", fontweight="bold")

plt.title("📏 Distance Between Two Points (Euclidean Distance)", fontsize=13, fontweight="bold")
plt.xlabel("X-axis", fontsize=11)
plt.ylabel("Y-axis", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(loc="upper left", frameon=True)
plt.tight_layout()
plt.show()
```

## 2️⃣ Straight Lines (Linear Equations)

### 📖 Concept (Simple English):
A straight line represents a constant rate of change between two variables ($x$ and $y$).

### 📐 Standard Equation (Slope-Intercept Form):
$$y = mx + c$$

- **$m$ (Slope):** The steepness/angle of the line. $m = \frac{\Delta y}{\Delta x} = \frac{y_2 - y_1}{x_2 - x_1} = \tan(\theta)$
- **$c$ (Y-intercept):** The value of $y$ where the line crosses the Y-axis ($x = 0$).

### 📐 General Linear Equation Form:
$$Ax + By + C = 0 \implies y = -\frac{A}{B}x - \frac{C}{B}$$
- Slope $m = -\frac{A}{B}$
- Y-intercept $c = -\frac{C}{B}$

### 🤖 Machine Learning Connection:
In **Simple Linear Regression**, the model hypothesis is a straight line: $\hat{y} = w_1 x + b$ where $w_1 = m$ (weight/slope) and $b = c$ (bias/intercept).

```python
# ==========================================================
# 2. STRAIGHT LINE: y = mx + c
# ==========================================================

x = np.linspace(-5, 5, 200)

# Different slopes (m) and intercepts (c)
m1, c1 = 2, 1     # Positive slope: y = 2x + 1
m2, c2 = -1, 3    # Negative slope: y = -x + 3
m3, c3 = 0, 2     # Zero slope (Horizontal line): y = 2

y1 = m1 * x + c1
y2 = m2 * x + c2
y3 = m3 * x + c3

# --- Visualization ---
plt.figure(figsize=(9, 6))

plt.plot(x, y1, color="blue", linewidth=2, label=f"Positive Slope: y = {m1}x + {c1}")
plt.plot(x, y2, color="crimson", linewidth=2, label=f"Negative Slope: y = {m2}x + {c2}")
plt.plot(x, y3, color="green", linewidth=2, linestyle="--", label=f"Zero Slope: y = {c3}")

# Axes and origin
plt.axhline(0, color="black", linewidth=1)
plt.axvline(0, color="black", linewidth=1)

plt.title("📈 Straight Lines with Different Slopes (y = mx + c)", fontsize=13, fontweight="bold")
plt.xlabel("X-axis", fontsize=11)
plt.ylabel("Y-axis", fontsize=11)
plt.ylim(-10, 10)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(loc="upper left", frameon=True)
plt.tight_layout()
plt.show()
```

## 3️⃣ Parallel Lines & Distance Between Parallel Lines

### 📖 Concept (Simple English):
- Two lines are **parallel** if they have the **exact same slope** ($m_1 = m_2$) but different Y-intercepts ($c_1 \neq c_2$).
- Parallel lines run in the same direction and **never intersect**.

### 📐 Equations of Two Parallel Lines:
1. Line 1: $y = mx + c_1$
2. Line 2: $y = mx + c_2$

### 📐 Perpendicular Distance Between Two Parallel Lines ($d$):
$$d = \frac{|c_1 - c_2|}{\sqrt{1 + m^2}}$$

Or in general form ($Ax + By + C_1 = 0$ and $Ax + By + C_2 = 0$):
$$d = \frac{|C_1 - C_2|}{\sqrt{A^2 + B^2}}$$

### 🤖 Machine Learning Connection (SVM Margin):
In **Support Vector Machines (SVM)**, the decision boundary is a hyperplane, and the margin between the two parallel support planes is calculated using this exact formula: $\text{Margin} = \frac{2}{\|w\|}$.

```python
# ==========================================================
# 3. PARALLEL LINES & PERPENDICULAR DISTANCE CALCULATION
# ==========================================================

# Slope and intercepts for two parallel lines
m = 1.5   # Same slope
c1 = 4    # Line 1 intercept
c2 = -2   # Line 2 intercept

# Distance formula between parallel lines
dist_parallel = abs(c1 - c2) / np.sqrt(1 + m**2)
print(f"Line 1 equation : y = {m}x + {c1}")
print(f"Line 2 equation : y = {m}x + ({c2})")
print(f"Slope (m1 = m2) : {m}")
print(f"Distance between parallel lines: {dist_parallel:.4f}")

# Coordinate points to draw perpendicular connecting segment
# Point on Line 1 at x = 0: (0, c1)
# The perpendicular slope is -1/m
# Normal unit vector perpendicular to the line: (-m, 1) / sqrt(1 + m^2)
x_p1, y_p1 = 0, c1
# Find perpendicular foot on Line 2
# Normal displacement magnitude = dist_parallel
perp_slope = -1 / m
# Perpendicular line through (0, c1): y - c1 = perp_slope * (x - 0) => y = perp_slope * x + c1
# Intersection with Line 2: m*x + c2 = perp_slope*x + c1 => x = (c1 - c2)/(m - perp_slope)
x_p2 = (c1 - c2) / (m - perp_slope)
y_p2 = m * x_p2 + c2

# --- Visualization ---
x_vals = np.linspace(-4, 5, 200)
y_line1 = m * x_vals + c1
y_line2 = m * x_vals + c2

plt.figure(figsize=(9, 6.5))
plt.plot(x_vals, y_line1, color="#1f77b4", linewidth=2.5, label=f"Line 1: y = {m}x + {c1}")
plt.plot(x_vals, y_line2, color="#ff7f0e", linewidth=2.5, label=f"Line 2: y = {m}x + ({c2})")

# Draw perpendicular distance line between them
plt.plot([x_p1, x_p2], [y_p1, y_p2], color="red", linewidth=2.5, linestyle="--", 
         label=f"Perpendicular Distance d = {dist_parallel:.2f}")
plt.scatter([x_p1, x_p2], [y_p1, y_p2], color="red", s=80, zorder=5)

# Annotations
plt.text(x_p1 - 0.7, y_p1 + 0.3, f"(0, {c1})", fontsize=10, fontweight="bold", color="#1f77b4")
plt.text(x_p2 + 0.2, y_p2 - 0.5, f"({x_p2:.2f}, {y_p2:.2f})", fontsize=10, fontweight="bold", color="#ff7f0e")
plt.text((x_p1 + x_p2)/2 + 0.2, (y_p1 + y_p2)/2, f"d = {dist_parallel:.2f}", fontsize=11, color="red", fontweight="bold")

plt.title(f"📏 Parallel Lines & Shortest Distance (d = |c1 - c2| / √(1 + m²))", fontsize=13, fontweight="bold")
plt.xlabel("X-axis", fontsize=11)
plt.ylabel("Y-axis", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(loc="upper left", frameon=True)
plt.tight_layout()
plt.show()
```

## 4️⃣ Perpendicular Lines (Orthogonal Lines)

### 📖 Concept (Simple English):
- Two lines are **perpendicular** if they intersect at a **$90^\circ$ right angle**.
- The slope of one line is the **negative reciprocal** of the other.

### 📐 Slopes Condition:
$$m_1 \cdot m_2 = -1 \implies m_2 = -\frac{1}{m_1}$$

### 🤖 Machine Learning Connection:
In Machine Learning and Linear Algebra, perpendicularity is known as **Orthogonality**:
- Two vectors are orthogonal if their dot product is zero: $\mathbf{u} \cdot \mathbf{v} = 0$.
- **Principal Component Analysis (PCA)** finds orthogonal axes to capture maximum variance without redundant information.

```python
# ==========================================================
# 4. PERPENDICULAR LINES: m1 * m2 = -1
# ==========================================================

# Line 1 parameters
m1 = 2.0
c1 = 1.0

# Line 2 parameters (Perpendicular slope: -1/m1)
m2 = -1 / m1
c2 = 4.0

print(f"Line 1 slope (m1)           : {m1}")
print(f"Line 2 slope (m2 = -1/m1)   : {m2}")
print(f"Product check (m1 * m2)     : {m1 * m2} (Must be -1)")

# Find intersection point (x_int, y_int)
# m1*x + c1 = m2*x + c2 => (m1 - m2)*x = c2 - c1
x_int = (c2 - c1) / (m1 - m2)
y_int = m1 * x_int + c1
print(f"Intersection Point (90°): ({x_int:.2f}, {y_int:.2f})")

# --- Visualization ---
x_vals = np.linspace(-4, 6, 200)
y_line1 = m1 * x_vals + c1
y_line2 = m2 * x_vals + c2

plt.figure(figsize=(8, 8))  # Square aspect ratio to see true 90-degree right angle
plt.plot(x_vals, y_line1, color="#2ca02c", linewidth=2.5, label=f"Line 1: y = {m1}x + {c1}")
plt.plot(x_vals, y_line2, color="#9467bd", linewidth=2.5, label=f"Line 2: y = {m2}x + {c2}")

# Intersection point
plt.scatter([x_int], [y_int], color="red", s=130, zorder=5, label=f"Intersection (90° Angle)")
plt.text(x_int + 0.3, y_int - 0.6, f"({x_int:.2f}, {y_int:.2f})\nAngle = 90°", fontsize=10, fontweight="bold")

# Equal aspect ratio is required to visually perceive the perpendicular 90 degree angle
plt.gca().set_aspect('equal', adjustable='box')
plt.xlim(-4, 6)
plt.ylim(-4, 6)
plt.title("📐 Perpendicular Lines (m1 · m2 = -1, Orthogonal at 90°)", fontsize=13, fontweight="bold")
plt.xlabel("X-axis", fontsize=11)
plt.ylabel("Y-axis", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(loc="upper left", frameon=True)
plt.tight_layout()
plt.show()
```

## 5️⃣ Vector Addition & Geometric Visualizations

### 📖 Concept (Simple English):
- A **vector** has both **magnitude (length)** and **direction**.
- When we add two vectors $\mathbf{u}$ and $\mathbf{v}$, we add their corresponding components:
  $$\mathbf{w} = \mathbf{u} + \mathbf{v} = \begin{bmatrix} u_1 \\ u_2 \end{bmatrix} + \begin{bmatrix} v_1 \\ v_2 \end{bmatrix} = \begin{bmatrix} u_1 + v_1 \\ u_2 + v_2 \end{bmatrix}$$

### 📐 Geometric Rules for Vector Addition:
1. **Head-to-Tail Rule (Triangle Law):** Place the tail (start) of Vector $\mathbf{v}$ at the head (tip) of Vector $\mathbf{u}$. The resultant vector $\mathbf{u} + \mathbf{v}$ points from the tail of $\mathbf{u}$ to the head of $\mathbf{v}$.
2. **Parallelogram Law:** If two vectors start from the same origin, they form adjacent sides of a parallelogram. The diagonal starting from the origin represents their sum $(\mathbf{u} + \mathbf{v})$.

### 🤖 Machine Learning Connection:
- In **Word Embeddings (Word2Vec)**, vector addition captures semantic relationships: $\vec{v}_{\text{King}} - \vec{v}_{\text{Man}} + \vec{v}_{\text{Woman}} \approx \vec{v}_{\text{Queen}}$.
- In **Neural Networks**, layer activations combine weighted feature inputs: $\mathbf{z} = W\mathbf{x} + \mathbf{b}$ (Matrix-vector product + bias vector addition).

```python
# ==========================================================
# 5. VECTOR ADDITION & GEOMETRIC VISUALIZATIONS
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt

# Define two 2D vectors
u = np.array([4, 2])
v = np.array([2, 5])

# Vector Addition (Resultant)
w = u + v

print(f"Vector u                : {u}")
print(f"Vector v                : {v}")
print(f"Resultant Vector (u + v): {w}")
print(f"Magnitude of u          : {np.linalg.norm(u):.2f}")
print(f"Magnitude of v          : {np.linalg.norm(v):.2f}")
print(f"Magnitude of (u + v)    : {np.linalg.norm(w):.2f}")

# --- 2-Panel Side-by-Side Visualization ---
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Panel 1: Head-to-Tail Rule (Triangle Law)
ax1 = axes[0]
ax1.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, color='#1f77b4', label=f'Vector u {tuple(u)}')
ax1.quiver(u[0], u[1], v[0], v[1], angles='xy', scale_units='xy', scale=1, color='#2ca02c', label=f'Vector v {tuple(v)} (at head of u)')
ax1.quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1, color='#d62728', label=f'Resultant u + v {tuple(w)}')

ax1.text(u[0]/2 - 0.4, u[1]/2 + 0.3, f'u {tuple(u)}', fontsize=11, color='#1f77b4', fontweight='bold')
ax1.text(u[0] + v[0]/2 + 0.2, u[1] + v[1]/2, f'v {tuple(v)}', fontsize=11, color='#2ca02c', fontweight='bold')
ax1.text(w[0]/2 - 0.7, w[1]/2 + 0.4, f'u + v {tuple(w)}', fontsize=11, color='#d62728', fontweight='bold')

ax1.set_xlim(-1, 8)
ax1.set_ylim(-1, 9)
ax1.set_title("🔹 Head-to-Tail Method (Triangle Law)", fontsize=12, fontweight="bold")
ax1.set_xlabel("X-axis", fontsize=11)
ax1.set_ylabel("Y-axis", fontsize=11)
ax1.grid(True, linestyle="--", alpha=0.4)
ax1.axhline(0, color='black', linewidth=0.8)
ax1.axvline(0, color='black', linewidth=0.8)
ax1.legend(loc="upper left", frameon=True)

# Panel 2: Parallelogram Law (Both from Origin)
ax2 = axes[1]
ax2.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, color='#1f77b4', label=f'Vector u {tuple(u)}')
ax2.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='#2ca02c', label=f'Vector v {tuple(v)}')

ax2.plot([u[0], w[0]], [u[1], w[1]], color='#2ca02c', linestyle='--', linewidth=1.8, alpha=0.7, label='Parallel to v')
ax2.plot([v[0], w[0]], [v[1], w[1]], color='#1f77b4', linestyle='--', linewidth=1.8, alpha=0.7, label='Parallel to u')
ax2.quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1, color='#d62728', label=f'Diagonal: u + v {tuple(w)}')

ax2.scatter([w[0]], [w[1]], color='#d62728', s=80, zorder=5)
ax2.text(w[0] + 0.2, w[1] - 0.2, f'({w[0]}, {w[1]})', fontsize=11, fontweight='bold', color='#d62728')

ax2.set_xlim(-1, 8)
ax2.set_ylim(-1, 9)
ax2.set_title("🔹 Parallelogram Law (Diagonal = Sum)", fontsize=12, fontweight="bold")
ax2.set_xlabel("X-axis", fontsize=11)
ax2.set_ylabel("Y-axis", fontsize=11)
ax2.grid(True, linestyle="--", alpha=0.4)
ax2.axhline(0, color='black', linewidth=0.8)
ax2.axvline(0, color='black', linewidth=0.8)
ax2.legend(loc="upper left", frameon=True)

plt.suptitle("🎯 Vector Addition Visualizations (u = [4, 2], v = [2, 5], u + v = [6, 7])", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

## 6️⃣ Vector Scalar Multiplication (Scaling in 3D)

### 📖 Concept (Simple English):
- **Scalar Multiplication** means multiplying each component of a vector by a real number (scalar $c$):
  $$c \cdot \mathbf{v} = c \cdot \begin{bmatrix} v_1 \\ v_2 \\ v_3 \end{bmatrix} = \begin{bmatrix} c v_1 \\ c v_2 \\ c v_3 \end{bmatrix}$$
- If $c > 1$: Vector stretches (magnitude increases).
- If $0 < c < 1$: Vector shrinks (magnitude decreases).
- If $c < 0$: Vector reverses direction.
- Direction remains on the same straight line (parallel/collinear).

```python
# ==========================================================
# 6. SCALAR MULTIPLICATION & 3D VISUALIZATION
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 1. Vector & Scalar
v = np.array([2, 1, 3])
scalar = 3
resultant = v * scalar  # [6, 3, 9]

print(f"Vector v                : {v}")
print(f"Scalar                  : {scalar}")
print(f"Resultant (v * scalar)  : {resultant}")
print(f"Original Magnitude      : {np.linalg.norm(v):.2f}")
print(f"Scaled Magnitude (x 3)  : {np.linalg.norm(resultant):.2f}")

# 2. 3D Visualization
fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection='3d')
origin = np.array([0, 0, 0])

# Plot Resultant Vector 3v
ax.quiver(origin[0], origin[1], origin[2], resultant[0], resultant[1], resultant[2],
          color='crimson', arrow_length_ratio=0.1, linewidth=3.5, label=f'Resultant 3v: {tuple(resultant)}')
# Plot Original Vector v
ax.quiver(origin[0], origin[1], origin[2], v[0], v[1], v[2],
          color='blue', arrow_length_ratio=0.15, linewidth=4, label=f'Original v: {tuple(v)}')

ax.scatter([v[0]], [v[1]], [v[2]], color='blue', s=60)
ax.scatter([resultant[0]], [resultant[1]], [resultant[2]], color='crimson', s=80)

ax.text(v[0], v[1], v[2] + 0.5, f'v {tuple(v)}', color='blue', fontsize=10, fontweight='bold')
ax.text(resultant[0], resultant[1], resultant[2] + 0.5, f'3v {tuple(resultant)}', color='crimson', fontsize=10, fontweight='bold')

ax.set_xlim([0, 7])
ax.set_ylim([0, 5])
ax.set_zlim([0, 10])

ax.set_xlabel('X-axis', fontweight='bold')
ax.set_ylabel('Y-axis', fontweight='bold')
ax.set_zlabel('Z-axis', fontweight='bold')
ax.set_title('🎯 3D Vector Scalar Multiplication: Resultant = 3 × v', fontsize=13, fontweight='bold')
ax.legend(loc='upper left')

plt.tight_layout()
plt.show()
```

## 7️⃣ Cross Product (Vector Product)

### 📖 Concept (Simple English):
- **Cross Product ($\mathbf{u} \times \mathbf{v}$)** is defined only for **3D vectors**.
- While the Dot Product gives a single number (scalar), the **Cross Product gives a NEW 3D VECTOR**.
- **Key Geometric Property:** The resultant vector $(\mathbf{u} \times \mathbf{v})$ is **perpendicular ($90^\circ$ orthogonal)** to BOTH original vectors $\mathbf{u}$ and $\mathbf{v}$!
- **Right-Hand Rule:** If you curl the fingers of your right hand from $\mathbf{u}$ to $\mathbf{v}$, your thumb points in the direction of $\mathbf{u} \times \mathbf{v}$.

### 📐 Formulas:
1. **Magnitude & Area:**
   $$\|\mathbf{u} \times \mathbf{v}\| = \|\mathbf{u}\| \|\mathbf{v}\| \sin(\theta)$$
   - The magnitude equals the **Area of the Parallelogram** formed by $\mathbf{u}$ and $\mathbf{v}$.
   - If vectors are parallel ($\theta = 0^\circ$), their cross product is $\mathbf{0}$.

2. **Component Formula (Determinant Form):**
   $$\mathbf{u} \times \mathbf{v} = \det \begin{bmatrix} \mathbf{i} & \mathbf{j} & \mathbf{k} \\ u_1 & u_2 & u_3 \\ v_1 & v_2 & v_3 \end{bmatrix} = \begin{bmatrix} u_2 v_3 - u_3 v_2 \\ u_3 v_1 - u_1 v_3 \\ u_1 v_2 - u_2 v_1 \end{bmatrix}$$

3. **Anti-commutative Property:**
   $$\mathbf{v} \times \mathbf{u} = -(\mathbf{u} \times \mathbf{v})$$

### 🤖 Machine Learning Connection:
- **3D Computer Vision & NeRFs:** Essential for calculating **Surface Normals (perpendicular vectors)** on 3D meshes and point clouds.
- **Robotics:** Calculates torque, angular momentum, and rotational motion.

## 8️⃣ Determinants of Matrices ($\det(A)$ or $|A|$)

### 📖 Concept (Simple English):
- **What is a Determinant?**
  A special scalar number calculated only from **square matrices** ($2 \times 2, 3 \times 3, \dots, n \times n$).
- **Geometric Meaning (Area & Volume Scaling Factor):**
  - When a matrix acts as a linear transformation, the absolute value of its determinant $|\det(A)|$ represents the factor by which **areas (in 2D)** or **volumes (in 3D)** are scaled!
  - A unit square with Area $= 1$ transforms into a parallelogram with $\text{Area} = |\det(A)|$.
- **Orientation (Sign of Determinant):**
  - If $\det(A) > 0$: Orientation is preserved (counter-clockwise stays counter-clockwise).
  - If $\det(A) < 0$: Orientation is flipped/mirrored (space is inverted).
- **Singular Matrix ($\det(A) = 0$):**
  - When $\det(A) = 0$, 2D space is squashed into a 1D line (or 3D space into a 2D plane), collapsing area to 0.
  - Because information is lost, **a matrix with $\det(A) = 0$ is SINGULAR and has NO INVERSE ($A^{-1}$ does not exist)**.

### 📐 Mathematical Formulas:
1. **$2 \times 2$ Matrix Determinant:**
   $$A = \begin{bmatrix} a & b \\ c & d \end{bmatrix} \implies \det(A) = ad - bc$$

2. **$3 \times 3$ Matrix Determinant (Laplace Expansion):**
   $$B = \begin{bmatrix} a & b & c \\ d & e & f \\ g & h & i \end{bmatrix} \implies \det(B) = a(ei - fh) - b(di - fg) + c(dh - eg)$$

3. **Key Properties:**
   - $\det(I) = 1$ (Identity matrix preserves area/volume).
   - $\det(A^T) = \det(A)$ (Transpose has same determinant).
   - $\det(A \cdot B) = \det(A) \cdot \det(B)$.
   - $\det(A^{-1}) = \frac{1}{\det(A)}$ (Inverse exists only if $\det(A) \neq 0$).

### 🤖 Machine Learning Applications:
- **Multivariate Gaussian Distribution:**
  $$p(\mathbf{x}) = \frac{1}{\sqrt{(2\pi)^k |\boldsymbol{\Sigma}|}} \exp\left(-\frac{1}{2}(\mathbf{x}-\boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{x}-\boldsymbol{\mu})\right)$$
  The determinant of covariance matrix $|\boldsymbol{\Sigma}|$ represents the total volume of data dispersion/uncertainty.
- **Normalizing Flows:** Generative models compute the Jacobian determinant $|\det(J)|$ to track probability density changes across layers.
- **PCA & Eigenvalues:** $\det(A) = \prod_{i=1}^n \lambda_i$ (product of eigenvalues).

```python
# ==========================================================
# 8. DETERMINANTS: COMPUTATION & GEOMETRIC AREA SCALING
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

# 1. 2x2 Matrix Determinant
A = np.array([[3, 1],
              [1, 2]])
det_A = np.linalg.det(A)

print("--- 2x2 Matrix A ---")
print(A)
print(f"Manual Formula (ad - bc) : (3*2 - 1*1) = {3*2 - 1*1}")
print(f"NumPy np.linalg.det(A)   : {det_A:.4f}")

# 2. 3x3 Matrix Determinant
B = np.array([[1, 2, 3],
              [0, 1, 4],
              [5, 6, 0]])
det_B = np.linalg.det(B)

print("\n--- 3x3 Matrix B ---")
print(B)
print(f"NumPy np.linalg.det(B)   : {det_B:.4f}")

# 3. Singular Matrix (det = 0)
C = np.array([[2, 4],
              [1, 2]])
det_C = np.linalg.det(C)
print("\n--- Singular Matrix C (Linearly Dependent Columns) ---")
print(C)
print(f"Determinant det(C)       : {det_C:.4f} (Singular! No inverse exists)")

# ==========================================================
# GEOMETRIC VISUALIZATION OF DETERMINANT AS AREA SCALING
# ==========================================================
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5.5))

# --- Panel 1: Original Unit Square ---
origin = np.array([0, 0])
i_hat = np.array([1, 0])
j_hat = np.array([0, 1])
square_pts = np.array([[0, 0], [1, 0], [1, 1], [0, 1]])

ax1.add_patch(Polygon(square_pts, closed=True, facecolor='#a6c8e0', edgecolor='#1f77b4', alpha=0.5, linewidth=2))
ax1.quiver(0, 0, i_hat[0], i_hat[1], angles='xy', scale_units='xy', scale=1,
           color='#1f77b4', linewidth=3, label='i_hat (1, 0)')
ax1.quiver(0, 0, j_hat[0], j_hat[1], angles='xy', scale_units='xy', scale=1,
           color='#2ca02c', linewidth=3, label='j_hat (0, 1)')
ax1.text(0.35, 0.45, "Area = 1", fontsize=12, fontweight='bold', color='#0b3c5d')

ax1.set_xlim([-0.5, 4.5])
ax1.set_ylim([-0.5, 4.5])
ax1.axhline(0, color='gray', linestyle='--', alpha=0.5)
ax1.axvline(0, color='gray', linestyle='--', alpha=0.5)
ax1.set_aspect('equal')
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.set_title("Original Space: Unit Square\nArea = 1.00", fontsize=12, fontweight='bold')
ax1.legend(loc='upper left')

# --- Panel 2: Transformed by Matrix A (det(A) = 5) ---
u = A @ i_hat    # [3, 1]
v = A @ j_hat    # [1, 2]
corner = u + v   # [4, 3]
para_pts = np.array([[0, 0], u, corner, v])

ax2.add_patch(Polygon(para_pts, closed=True, facecolor='#fde3a7', edgecolor='#f39c12', alpha=0.6, linewidth=2))
ax2.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1,
           color='#1f77b4', linewidth=3, label=f'A·i_hat {tuple(u)}')
ax2.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
           color='#2ca02c', linewidth=3, label=f'A·j_hat {tuple(v)}')
ax2.text(corner[0]/2 - 0.4, corner[1]/2, f"Area = |det(A)|\n       = {abs(det_A):.1f}",
         fontsize=12, fontweight='bold', color='#b76e00')

ax2.set_xlim([-0.5, 4.5])
ax2.set_ylim([-0.5, 4.5])
ax2.axhline(0, color='gray', linestyle='--', alpha=0.5)
ax2.axvline(0, color='gray', linestyle='--', alpha=0.5)
ax2.set_aspect('equal')
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.set_title(f"Transformed by Matrix A\nArea Scaled by det(A) = {det_A:.0f}x", fontsize=12, fontweight='bold')
ax2.legend(loc='upper left')

# --- Panel 3: Singular Matrix C (det(C) = 0, Squashed to 1D Line) ---
u_sing = C @ i_hat  # [2, 1]
v_sing = C @ j_hat  # [4, 2]
ax3.quiver(0, 0, v_sing[0], v_sing[1], angles='xy', scale_units='xy', scale=1,
           color='#e74c3c', linewidth=4, label=f'C·j_hat {tuple(v_sing)}')
ax3.quiver(0, 0, u_sing[0], u_sing[1], angles='xy', scale_units='xy', scale=1,
           color='#8e44ad', linewidth=4, label=f'C·i_hat {tuple(u_sing)}')
ax3.plot([0, 5], [0, 2.5], color='red', linestyle='--', alpha=0.7)
ax3.text(1.0, 1.5, "Area = det(C) = 0\n(Squashed into 1D Line!\nNo Inverse Exists)",
         fontsize=11, fontweight='bold', color='#c0392b')

ax3.set_xlim([-0.5, 4.5])
ax3.set_ylim([-0.5, 4.5])
ax3.axhline(0, color='gray', linestyle='--', alpha=0.5)
ax3.axvline(0, color='gray', linestyle='--', alpha=0.5)
ax3.set_aspect('equal')
ax3.grid(True, linestyle=':', alpha=0.6)
ax3.set_title("Singular Matrix C (det = 0)\n2D Area Collapsed to 0", fontsize=12, fontweight='bold')
ax3.legend(loc='upper left')

plt.tight_layout()
plt.show()
```

## 9️⃣ Matrix Operations & Matrix Inverse (Linear Systems & Foundations)

### 📖 Concept (Simple English):
- **Dot Product ($\mathbf{u} \cdot \mathbf{v}$):** Measures similarity and projection between vectors. When $\mathbf{u} \cdot \mathbf{v} = 0$, the vectors are **orthogonal ($90^\circ$ perpendicular)**.
- **Matrix Inverse ($A^{-1}$):** A matrix that reverses the transformation of $A$:
  $$A \cdot A^{-1} = A^{-1} \cdot A = I \quad \text{(Identity Matrix)}$$
- **Connection to Determinants:**
  $$A^{-1} = \frac{1}{\det(A)} \operatorname{adj}(A)$$
  - Because $\det(A)$ is in the denominator, if $\det(A) = 0$, dividing by zero is undefined!
  - Therefore, **only matrices with $\det(A) \neq 0$ can be inverted**.

### 🤖 Machine Learning Applications:
- **Linear Regression (Normal Equation):**
  $$\mathbf{w} = (X^T X)^{-1} X^T \mathbf{y}$$
  Requires $(X^T X)$ to be non-singular (invertible). If features are multicollinear, $\det(X^T X) \approx 0$, which causes numerical instability (solved by Ridge Regularization!).
- **Neural Networks:** Weight matrices, affine transformations $W\mathbf{x} + \mathbf{b}$, and backpropagation.

```python
# ==========================================================
# 9. VECTORS, DOT PRODUCT & MATRIX INVERSE
# ==========================================================
import numpy as np

# --- 1. Vectors & Dot Product ---
u = np.array([3, 4])
v = np.array([4, -3])

dot_prod = np.dot(u, v)
print(f"Vector u: {u}")
print(f"Vector v: {v}")
print(f"Dot Product u · v: {dot_prod}")
if dot_prod == 0:
    print("✅ u and v are ORTHOGONAL (Perpendicular) vectors because u · v = 0!")

# --- 2. Invertible Matrix A ---
A = np.array([[2, 1],
              [1, 3]])

det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)

print("\nMatrix A:")
print(A)
print(f"Determinant det(A): {det_A:.2f} (Non-zero --> Invertible!)")
print("Inverse Matrix A^(-1):")
print(inv_A)

# Verification: A @ A_inv = Identity Matrix I
I = np.round(A @ inv_A, decimals=5)
print("Verification (A @ A^(-1) = Identity):")
print(I)

# --- 3. Singular Matrix Inversion Attempt ---
C = np.array([[2, 4],
              [1, 2]])
print(f"\nSingular Matrix C:\n{C}")
print(f"det(C) = {np.linalg.det(C):.4f}")
try:
    np.linalg.inv(C)
except np.linalg.LinAlgError as e:
    print(f"❌ Inversion Failed as expected: {e}")
    print("💡 Reason: det(C) = 0 means matrix is Singular (columns are linearly dependent)!")
```

## 🔟 Vector Projection (Geometric Shadow & Orthogonality)

### 📖 Concept (Simple English):
- **Vector Projection:** Projecting vector $\mathbf{v}$ onto vector $\mathbf{u}$ is like shining a light directly perpendicular to $\mathbf{u}$ and looking at the **shadow** that $\mathbf{v}$ casts along $\mathbf{u}$.
- **Scalar Projection (Length/Magnitude of shadow):**
  $$\text{comp}_{\mathbf{u}}(\mathbf{v}) = \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\|} = \|\mathbf{v}\| \cos(\theta)$$
- **Vector Projection (Actual vector along $\mathbf{u}$):**
  $$\operatorname{proj}_{\mathbf{u}}(\mathbf{v}) = \left(\frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\|^2}\right) \mathbf{u}$$
- **Orthogonal Component (Perpendicular error):**
  $$\mathbf{v}_{\perp} = \mathbf{v} - \operatorname{proj}_{\mathbf{u}}(\mathbf{v})$$
  This perpendicular component has $\mathbf{v}_{\perp} \cdot \mathbf{u} = 0$.

### 🤖 Machine Learning Applications:
- **Principal Component Analysis (PCA):** Projecting high-dimensional data points onto the principal component axes to reduce dimensions.
- **Cosine Similarity:** Used in NLP and recommendation systems to measure angle between embeddings ($\cos\theta = \frac{\mathbf{u}\cdot\mathbf{v}}{\|\mathbf{u}\|\|\mathbf{v}\|}$).
- **Support Vector Machines (SVM):** Distance of data points from the separating hyperplane is computed via vector projection.

```python
# ==========================================================
# 10. VECTOR PROJECTION: FORMULA & GEOMETRIC VISUALIZATION
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt

# 1. Define two vectors
u = np.array([5, 1])  # Base vector onto which we project
v = np.array([3, 4])  # Vector being projected

# 2. Vector Projection Calculation
# proj_u(v) = (u · v / ||u||^2) * u
u_dot_v = np.dot(u, v)
u_norm_sq = np.dot(u, u)
scalar_factor = u_dot_v / u_norm_sq
proj_v_on_u = scalar_factor * u

# Perpendicular component (error vector)
perp_component = v - proj_v_on_u

print(f"Base Vector u               : {u}")
print(f"Vector v                    : {v}")
print(f"Dot Product (u · v)         : {u_dot_v}")
print(f"Vector Projection proj_u(v) : {proj_v_on_u}")
print(f"Scalar Projection (Length)  : {np.linalg.norm(proj_v_on_u):.4f}")
print(f"Perpendicular Component     : {perp_component}")
print(f"Orthogonality Check (perp · u): {np.dot(perp_component, u):.6f} (Strictly 0!)")

# 3. 2D Visualization
plt.figure(figsize=(9, 7))

# Draw line along vector u
line_u = np.linspace(-1, 7, 100)
slope_u = u[1] / u[0]
plt.plot(line_u, slope_u * line_u, color='gray', linestyle=':', alpha=0.5, label='Direction of u')

# Plot Base Vector u
plt.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1,
           color='#1f77b4', linewidth=3, label=f'Base Vector u {tuple(u)}')

# Plot Vector v
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1,
           color='#2ca02c', linewidth=3, label=f'Vector v {tuple(v)}')

# Plot Projection Vector proj_u(v)
plt.quiver(0, 0, proj_v_on_u[0], proj_v_on_u[1], angles='xy', scale_units='xy', scale=1,
           color='#d62728', linewidth=4, label=f'proj_u(v): {tuple(np.round(proj_v_on_u, 2))}')

# Draw Perpendicular Drop Line (from tip of v to tip of projection)
plt.plot([v[0], proj_v_on_u[0]], [v[1], proj_v_on_u[1]],
         color='purple', linestyle='--', linewidth=2, label='Perpendicular Drop (v_perp)')

# Mark points
plt.scatter([v[0], proj_v_on_u[0]], [v[1], proj_v_on_u[1]], color=['#2ca02c', '#d62728'], s=60, zorder=5)
plt.text(v[0] + 0.1, v[1] + 0.2, f'v {tuple(v)}', fontweight='bold', color='#2ca02c', fontsize=11)
plt.text(proj_v_on_u[0] + 0.1, proj_v_on_u[1] - 0.35, f'proj_u(v)', fontweight='bold', color='#d62728', fontsize=11)

plt.xlim([-1, 7])
plt.ylim([-1, 6])
plt.axhline(0, color='black', linewidth=0.8, alpha=0.6)
plt.axvline(0, color='black', linewidth=0.8, alpha=0.6)
plt.gca().set_aspect('equal')
plt.grid(True, linestyle=':', alpha=0.6)
plt.title('📐 Vector Projection: proj_u(v) as Shadow of v onto u', fontsize=13, fontweight='bold')
plt.legend(loc='upper left')

plt.tight_layout()
plt.show()
```

## 1️⃣1️⃣ Eigenvalues & Eigenvectors ($A\mathbf{v} = \lambda \mathbf{v}$)

### 📖 Concept (Simple English):
- **Normal Vectors Under Matrix Transformation:**
  When a matrix multiplies a regular vector, the vector usually changes **both its direction and length**.
- **Eigenvectors (The Invariant Vectors):**
  An **Eigenvector** is a special vector whose **direction DOES NOT CHANGE** after matrix transformation! It stays on the exact same line.
- **Eigenvalue ($\lambda$):**
  The scalar factor by which the eigenvector is stretched, shrunk, or reversed:
  $$A \mathbf{v} = \lambda \mathbf{v}$$
  - If $\lambda > 1$: Eigenvector stretches.
  - If $0 < \lambda < 1$: Eigenvector shrinks.
  - If $\lambda < 0$: Eigenvector flips $180^\circ$ direction.

### 📐 Characteristic Equation:
$$(A - \lambda I)\mathbf{v} = \mathbf{0}$$
For a non-zero eigenvector $\mathbf{v}$ to exist, the matrix $(A - \lambda I)$ must be singular:
$$\det(A - \lambda I) = 0$$

### 🤖 Machine Learning Applications:
- **Principal Component Analysis (PCA):** The eigenvectors of the feature covariance matrix define the **Principal Components** (directions of maximum data variance), and the eigenvalues represent the amount of variance explained!
- **Google PageRank:** The stationary probability distribution of web surfing is the principal eigenvector ($\lambda = 1$) of the Google Markov matrix.
- **Image Compression & SVD:** Singular Value Decomposition ($A = U \Sigma V^T$) is directly derived from eigenvalues and eigenvectors of $A A^T$ and $A^T A$.

```python
# ==========================================================
# 11. EIGENVALUES & EIGENVECTORS: NUMPY & 2D VISUALIZATION
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt

# 1. Define a 2x2 Transformation Matrix
A = np.array([[2, 1],
              [1, 2]])

# 2. Compute Eigenvalues & Eigenvectors using np.linalg.eig
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix A:")
print(A)
print(f"\nEigenvalues (λ): {eigenvalues}")
print("\nEigenvectors (columns of matrix):")
print(eigenvectors)

# Verification: A @ v = λ * v for each eigenvector
for i in range(len(eigenvalues)):
    lam = eigenvalues[i]
    v = eigenvectors[:, i]
    Av = A @ v
    lam_v = lam * v
    print(f"\n--- Verification for Eigenvector {i+1} ---")
    print(f"v_{i+1}          : {v}")
    print(f"A @ v_{i+1}      : {Av}")
    print(f"λ_{i+1} * v_{i+1}    : {lam_v}")
    print(f"Matches?       : {np.allclose(Av, lam_v)} ✅")

# 3. Geometric Visualization of Eigenvectors
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

v1 = eigenvectors[:, 0]
v2 = eigenvectors[:, 1]
lam1 = eigenvalues[0]
lam2 = eigenvalues[1]

# Arbitrary non-eigenvector for comparison
v_rand = np.array([1.0, 0.2])
Av_rand = A @ v_rand

# --- Plot 1: Before Transformation ---
ax1.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1,
           color='#1f77b4', linewidth=3.5, label=f'Eigenvector 1 {tuple(np.round(v1, 2))}')
ax1.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1,
           color='#2ca02c', linewidth=3.5, label=f'Eigenvector 2 {tuple(np.round(v2, 2))}')
ax1.quiver(0, 0, v_rand[0], v_rand[1], angles='xy', scale_units='xy', scale=1,
           color='#e74c3c', linewidth=3, linestyle='dashed', label=f'Regular vector {tuple(v_rand)}')

ax1.set_xlim([-4, 4])
ax1.set_ylim([-4, 4])
ax1.axhline(0, color='gray', linestyle='--', alpha=0.5)
ax1.axvline(0, color='gray', linestyle='--', alpha=0.5)
ax1.set_aspect('equal')
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.set_title("Original Vectors (Before Transformation)", fontsize=12, fontweight='bold')
ax1.legend(loc='upper left')

# --- Plot 2: After Transformation (A @ v) ---
# Transformed Eigenvectors stay on the same line, just stretched by lambda!
ax2.quiver(0, 0, lam1 * v1[0], lam1 * v1[1], angles='xy', scale_units='xy', scale=1,
           color='#1f77b4', linewidth=3.5, label=f'A·v1 = {lam1:.0f}·v1 (Direction Unchanged!)')
ax2.quiver(0, 0, lam2 * v2[0], lam2 * v2[1], angles='xy', scale_units='xy', scale=1,
           color='#2ca02c', linewidth=3.5, label=f'A·v2 = {lam2:.0f}·v2 (Direction Unchanged!)')
ax2.quiver(0, 0, Av_rand[0], Av_rand[1], angles='xy', scale_units='xy', scale=1,
           color='#e74c3c', linewidth=3, linestyle='dashed', label=f'A·v_rand (Direction Changed!)')

# Show invariant lines
line_x = np.linspace(-3, 3, 50)
ax2.plot(line_x, (v1[1]/v1[0]) * line_x, color='#1f77b4', linestyle=':', alpha=0.4)
ax2.plot(line_x, (v2[1]/v2[0]) * line_x, color='#2ca02c', linestyle=':', alpha=0.4)

ax2.set_xlim([-4, 4])
ax2.set_ylim([-4, 4])
ax2.axhline(0, color='gray', linestyle='--', alpha=0.5)
ax2.axvline(0, color='gray', linestyle='--', alpha=0.5)
ax2.set_aspect('equal')
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.set_title(f"After Matrix Transformation A\nEigenvectors scaled by λ1={lam1:.0f}, λ2={lam2:.0f}", fontsize=12, fontweight='bold')
ax2.legend(loc='upper left')

plt.tight_layout()
plt.show()
```
