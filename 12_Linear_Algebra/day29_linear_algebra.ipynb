## 1️⃣ Scalars, Vectors, Matrices & Tensors

- **Scalar (0D Tensor):** Single number (e.g., $x = 5$)
- **Vector (1D Tensor):** 1D array of numbers with magnitude & direction (e.g., $[1, 2, 3]$)
- **Matrix (2D Tensor):** 2D grid/table of numbers (Rows $\times$ Columns)
- **Tensor (3D+ Tensor):** Multi-dimensional array (e.g., RGB Image: Height $\times$ Width $\times$ 3)

```python
# Scalar (0D)
scalar = np.array(5)
print(f"Scalar: {scalar}, Shape: {scalar.shape}, Dimension: {scalar.ndim}D")

# Vector (1D)
vector = np.array([1, 2, 3])
print(f"Vector: {vector}, Shape: {vector.shape}, Dimension: {vector.ndim}D")

# Matrix (2D)
matrix = np.array([[1, 2], [3, 4]])
print(f"Matrix:\n{matrix}\nShape: {matrix.shape}, Dimension: {matrix.ndim}D")

# Tensor (3D)
tensor = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(f"Tensor:\n{tensor}\nShape: {tensor.shape}, Dimension: {tensor.ndim}D")
```

## 2️⃣ Vector Operations & Dot Product

### Dot Product Formula:
$$\mathbf{u} \cdot \mathbf{v} = \sum_{i=1}^{n} u_i v_i = u_1 v_1 + u_2 v_2 + \dots + u_n v_n$$

Machine learning mein weights ($W$) aur features ($X$) ka prediction calculation dot product se hota hai: $\hat{y} = W \cdot X + b$

```python
u = np.array([2, 3])
v = np.array([4, 1])

# Vector Addition & Subtraction
print(f"u + v = {u + v}")
print(f"u - v = {u - v}")

# Dot Product
dot_product = np.dot(u, v)  # ya u @ v
print(f"Dot Product (u . v) = {dot_product}")
```

## 3️⃣ Matrix Multiplication & Inverse

Matrix multiplication ($A \times B$) tabhi mumkin hai jab **A ke columns = B ki rows** hon.

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Matrix Multiplication
C = np.dot(A, B)  # ya A @ B
print(f"A @ B =\n{C}")

# Transpose (Rows -> Columns)
print(f"A Transpose:\n{A.T}")

# Determinant & Inverse
det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)
print(f"Determinant of A: {det_A:.2f}")
print(f"Inverse of A:\n{inv_A}")

# Verification: A @ A_inv = Identity Matrix
identity = np.round(A @ inv_A)
print(f"Verification (A @ A_inv):\n{identity}")
```
