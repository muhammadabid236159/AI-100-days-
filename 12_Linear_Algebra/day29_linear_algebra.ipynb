## 1️⃣ Distance Between Two Points

### 📖 Concept (Simple English):
The distance between two points in a 2D plane is the shortest straight-line length connecting them. It is calculated using the Pythagorean Theorem.

- **Point 1:** $P_1 = (x_1, y_1)$
- **Point 2:** $P_2 = (x_2, y_2)$

### 📐 Formula (Euclidean Distance / $L_2$ Norm):
$$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

### 🤖 Machine Learning Connection:
Euclidean distance is the fundamental metric used in **K-Nearest Neighbors (KNN)** classification, **K-Means Clustering**, and error loss calculations.

```python
# ==========================================================
# 1. DISTANCE BETWEEN TWO POINTS & VISUALIZATION
# ==========================================================

# Define two points
p1 = np.array([2, 3])
p2 = np.array([8, 11])

# Calculate Euclidean Distance using formula & NumPy
distance_formula = np.sqrt((p2[0] - p1[0])**2 + (p2[1] - p1[1])**2)
distance_numpy = np.linalg.norm(p2 - p1)

print(f"Point 1 (x1, y1)   : {tuple(p1)}")
print(f"Point 2 (x2, y2)   : {tuple(p2)}")
print(f"Distance (Formula) : {distance_formula:.4f}")
print(f"Distance (np.norm) : {distance_numpy:.4f}")

# --- Visualization ---
plt.figure(figsize=(8, 6))

# Plot the points
plt.scatter([p1[0], p2[0]], [p1[1], p2[1]], color="red", s=120, zorder=5, label="Points P1 & P2")

# Plot the direct distance line
plt.plot([p1[0], p2[0]], [p1[1], p2[1]], color="#2b5c8f", linewidth=2.5, linestyle="-", label=f"Distance d = {distance_numpy:.2f}")

# Plot right-angled triangle components (dx, dy)
plt.plot([p1[0], p2[0]], [p1[1], p1[1]], color="gray", linestyle="--", alpha=0.7, label=f"Δx = {p2[0] - p1[0]}")
plt.plot([p2[0], p2[0]], [p1[1], p2[1]], color="gray", linestyle=":", alpha=0.7, label=f"Δy = {p2[1] - p1[1]}")

# Annotations
plt.text(p1[0] - 0.4, p1[1] + 0.4, f"P1 (2, 3)", fontsize=11, fontweight="bold")
plt.text(p2[0] + 0.2, p2[1], f"P2 (8, 11)", fontsize=11, fontweight="bold")
plt.text((p1[0] + p2[0])/2 - 0.5, (p1[1] + p2[1])/2 + 0.6, f"d = {distance_numpy:.2f}", fontsize=11, color="#2b5c8f", fontweight="bold")

plt.title("📏 Distance Between Two Points (Euclidean Distance)", fontsize=13, fontweight="bold")
plt.xlabel("X-axis", fontsize=11)
plt.ylabel("Y-axis", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(loc="upper left", frameon=True)
plt.tight_layout()
plt.show()
```

## 2️⃣ Straight Lines (Linear Equations)

### 📖 Concept (Simple English):
A straight line represents a constant rate of change between two variables ($x$ and $y$).

### 📐 Standard Equation (Slope-Intercept Form):
$$y = mx + c$$

- **$m$ (Slope):** The steepness/angle of the line. $m = \frac{\Delta y}{\Delta x} = \frac{y_2 - y_1}{x_2 - x_1} = \tan(\theta)$
- **$c$ (Y-intercept):** The value of $y$ where the line crosses the Y-axis ($x = 0$).

### 📐 General Linear Equation Form:
$$Ax + By + C = 0 \implies y = -\frac{A}{B}x - \frac{C}{B}$$
- Slope $m = -\frac{A}{B}$
- Y-intercept $c = -\frac{C}{B}$

### 🤖 Machine Learning Connection:
In **Simple Linear Regression**, the model hypothesis is a straight line: $\hat{y} = w_1 x + b$ where $w_1 = m$ (weight/slope) and $b = c$ (bias/intercept).

```python
# ==========================================================
# 2. STRAIGHT LINE: y = mx + c
# ==========================================================

x = np.linspace(-5, 5, 200)

# Different slopes (m) and intercepts (c)
m1, c1 = 2, 1     # Positive slope: y = 2x + 1
m2, c2 = -1, 3    # Negative slope: y = -x + 3
m3, c3 = 0, 2     # Zero slope (Horizontal line): y = 2

y1 = m1 * x + c1
y2 = m2 * x + c2
y3 = m3 * x + c3

# --- Visualization ---
plt.figure(figsize=(9, 6))

plt.plot(x, y1, color="blue", linewidth=2, label=f"Positive Slope: y = {m1}x + {c1}")
plt.plot(x, y2, color="crimson", linewidth=2, label=f"Negative Slope: y = {m2}x + {c2}")
plt.plot(x, y3, color="green", linewidth=2, linestyle="--", label=f"Zero Slope: y = {c3}")

# Axes and origin
plt.axhline(0, color="black", linewidth=1)
plt.axvline(0, color="black", linewidth=1)

plt.title("📈 Straight Lines with Different Slopes (y = mx + c)", fontsize=13, fontweight="bold")
plt.xlabel("X-axis", fontsize=11)
plt.ylabel("Y-axis", fontsize=11)
plt.ylim(-10, 10)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(loc="upper left", frameon=True)
plt.tight_layout()
plt.show()
```

## 3️⃣ Parallel Lines & Distance Between Parallel Lines

### 📖 Concept (Simple English):
- Two lines are **parallel** if they have the **exact same slope** ($m_1 = m_2$) but different Y-intercepts ($c_1 \neq c_2$).
- Parallel lines run in the same direction and **never intersect**.

### 📐 Equations of Two Parallel Lines:
1. Line 1: $y = mx + c_1$
2. Line 2: $y = mx + c_2$

### 📐 Perpendicular Distance Between Two Parallel Lines ($d$):
$$d = \frac{|c_1 - c_2|}{\sqrt{1 + m^2}}$$

Or in general form ($Ax + By + C_1 = 0$ and $Ax + By + C_2 = 0$):
$$d = \frac{|C_1 - C_2|}{\sqrt{A^2 + B^2}}$$

### 🤖 Machine Learning Connection (SVM Margin):
In **Support Vector Machines (SVM)**, the decision boundary is a hyperplane, and the margin between the two parallel support planes is calculated using this exact formula: $\text{Margin} = \frac{2}{\|w\|}$.

```python
# ==========================================================
# 3. PARALLEL LINES & PERPENDICULAR DISTANCE CALCULATION
# ==========================================================

# Slope and intercepts for two parallel lines
m = 1.5   # Same slope
c1 = 4    # Line 1 intercept
c2 = -2   # Line 2 intercept

# Distance formula between parallel lines
dist_parallel = abs(c1 - c2) / np.sqrt(1 + m**2)
print(f"Line 1 equation : y = {m}x + {c1}")
print(f"Line 2 equation : y = {m}x + ({c2})")
print(f"Slope (m1 = m2) : {m}")
print(f"Distance between parallel lines: {dist_parallel:.4f}")

# Coordinate points to draw perpendicular connecting segment
# Point on Line 1 at x = 0: (0, c1)
# The perpendicular slope is -1/m
# Normal unit vector perpendicular to the line: (-m, 1) / sqrt(1 + m^2)
x_p1, y_p1 = 0, c1
# Find perpendicular foot on Line 2
# Normal displacement magnitude = dist_parallel
perp_slope = -1 / m
# Perpendicular line through (0, c1): y - c1 = perp_slope * (x - 0) => y = perp_slope * x + c1
# Intersection with Line 2: m*x + c2 = perp_slope*x + c1 => x = (c1 - c2)/(m - perp_slope)
x_p2 = (c1 - c2) / (m - perp_slope)
y_p2 = m * x_p2 + c2

# --- Visualization ---
x_vals = np.linspace(-4, 5, 200)
y_line1 = m * x_vals + c1
y_line2 = m * x_vals + c2

plt.figure(figsize=(9, 6.5))
plt.plot(x_vals, y_line1, color="#1f77b4", linewidth=2.5, label=f"Line 1: y = {m}x + {c1}")
plt.plot(x_vals, y_line2, color="#ff7f0e", linewidth=2.5, label=f"Line 2: y = {m}x + ({c2})")

# Draw perpendicular distance line between them
plt.plot([x_p1, x_p2], [y_p1, y_p2], color="red", linewidth=2.5, linestyle="--", 
         label=f"Perpendicular Distance d = {dist_parallel:.2f}")
plt.scatter([x_p1, x_p2], [y_p1, y_p2], color="red", s=80, zorder=5)

# Annotations
plt.text(x_p1 - 0.7, y_p1 + 0.3, f"(0, {c1})", fontsize=10, fontweight="bold", color="#1f77b4")
plt.text(x_p2 + 0.2, y_p2 - 0.5, f"({x_p2:.2f}, {y_p2:.2f})", fontsize=10, fontweight="bold", color="#ff7f0e")
plt.text((x_p1 + x_p2)/2 + 0.2, (y_p1 + y_p2)/2, f"d = {dist_parallel:.2f}", fontsize=11, color="red", fontweight="bold")

plt.title(f"📏 Parallel Lines & Shortest Distance (d = |c1 - c2| / √(1 + m²))", fontsize=13, fontweight="bold")
plt.xlabel("X-axis", fontsize=11)
plt.ylabel("Y-axis", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(loc="upper left", frameon=True)
plt.tight_layout()
plt.show()
```

## 4️⃣ Perpendicular Lines (Orthogonal Lines)

### 📖 Concept (Simple English):
- Two lines are **perpendicular** if they intersect at a **$90^\circ$ right angle**.
- The slope of one line is the **negative reciprocal** of the other.

### 📐 Slopes Condition:
$$m_1 \cdot m_2 = -1 \implies m_2 = -\frac{1}{m_1}$$

### 🤖 Machine Learning Connection:
In Machine Learning and Linear Algebra, perpendicularity is known as **Orthogonality**:
- Two vectors are orthogonal if their dot product is zero: $\mathbf{u} \cdot \mathbf{v} = 0$.
- **Principal Component Analysis (PCA)** finds orthogonal axes to capture maximum variance without redundant information.

```python
# ==========================================================
# 4. PERPENDICULAR LINES: m1 * m2 = -1
# ==========================================================

# Line 1 parameters
m1 = 2.0
c1 = 1.0

# Line 2 parameters (Perpendicular slope: -1/m1)
m2 = -1 / m1
c2 = 4.0

print(f"Line 1 slope (m1)           : {m1}")
print(f"Line 2 slope (m2 = -1/m1)   : {m2}")
print(f"Product check (m1 * m2)     : {m1 * m2} (Must be -1)")

# Find intersection point (x_int, y_int)
# m1*x + c1 = m2*x + c2 => (m1 - m2)*x = c2 - c1
x_int = (c2 - c1) / (m1 - m2)
y_int = m1 * x_int + c1
print(f"Intersection Point (90°): ({x_int:.2f}, {y_int:.2f})")

# --- Visualization ---
x_vals = np.linspace(-4, 6, 200)
y_line1 = m1 * x_vals + c1
y_line2 = m2 * x_vals + c2

plt.figure(figsize=(8, 8))  # Square aspect ratio to see true 90-degree right angle
plt.plot(x_vals, y_line1, color="#2ca02c", linewidth=2.5, label=f"Line 1: y = {m1}x + {c1}")
plt.plot(x_vals, y_line2, color="#9467bd", linewidth=2.5, label=f"Line 2: y = {m2}x + {c2}")

# Intersection point
plt.scatter([x_int], [y_int], color="red", s=130, zorder=5, label=f"Intersection (90° Angle)")
plt.text(x_int + 0.3, y_int - 0.6, f"({x_int:.2f}, {y_int:.2f})\nAngle = 90°", fontsize=10, fontweight="bold")

# Equal aspect ratio is required to visually perceive the perpendicular 90 degree angle
plt.gca().set_aspect('equal', adjustable='box')
plt.xlim(-4, 6)
plt.ylim(-4, 6)
plt.title("📐 Perpendicular Lines (m1 · m2 = -1, Orthogonal at 90°)", fontsize=13, fontweight="bold")
plt.xlabel("X-axis", fontsize=11)
plt.ylabel("Y-axis", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(loc="upper left", frameon=True)
plt.tight_layout()
plt.show()
```

## 5️⃣ Vector Addition & Geometric Visualizations

### 📖 Concept (Simple English):
- A **vector** has both **magnitude (length)** and **direction**.
- When we add two vectors $\mathbf{u}$ and $\mathbf{v}$, we add their corresponding components:
  $$\mathbf{w} = \mathbf{u} + \mathbf{v} = \begin{bmatrix} u_1 \\ u_2 \end{bmatrix} + \begin{bmatrix} v_1 \\ v_2 \end{bmatrix} = \begin{bmatrix} u_1 + v_1 \\ u_2 + v_2 \end{bmatrix}$$

### 📐 Geometric Rules for Vector Addition:
1. **Head-to-Tail Rule (Triangle Law):** Place the tail (start) of Vector $\mathbf{v}$ at the head (tip) of Vector $\mathbf{u}$. The resultant vector $\mathbf{u} + \mathbf{v}$ points from the tail of $\mathbf{u}$ to the head of $\mathbf{v}$.
2. **Parallelogram Law:** If two vectors start from the same origin, they form adjacent sides of a parallelogram. The diagonal starting from the origin represents their sum $(\mathbf{u} + \mathbf{v})$.

### 🤖 Machine Learning Connection:
- In **Word Embeddings (Word2Vec)**, vector addition captures semantic relationships: $\vec{v}_{\text{King}} - \vec{v}_{\text{Man}} + \vec{v}_{\text{Woman}} \approx \vec{v}_{\text{Queen}}$.
- In **Neural Networks**, layer activations combine weighted feature inputs: $\mathbf{z} = W\mathbf{x} + \mathbf{b}$ (Matrix-vector product + bias vector addition).

```python
# ==========================================================
# 5. VECTOR ADDITION & GEOMETRIC VISUALIZATIONS
# ==========================================================

# Define two 2D vectors
u = np.array([4, 2])
v = np.array([2, 5])

# Vector Addition (Resultant)
w = u + v

print(f"Vector u               : {u}")
print(f"Vector v               : {v}")
print(f"Resultant Vector (u + v): {w}")
print(f"Magnitude of u         : {np.linalg.norm(u):.2f}")
print(f"Magnitude of v         : {np.linalg.norm(v):.2f}")
print(f"Magnitude of (u + v)   : {np.linalg.norm(w):.2f}")

# --- 2-Panel Side-by-Side Visualization ---
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# ----------------------------------------------------------
# Subplot 1: Head-to-Tail Rule (Triangle Law)
# ----------------------------------------------------------
ax1 = axes[0]

# Plot Vector u from origin (0, 0)
ax1.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, color='#1f77b4', label=f'Vector u {tuple(u)}')
# Plot Vector v placed at head of u (u[0], u[1])
ax1.quiver(u[0], u[1], v[0], v[1], angles='xy', scale_units='xy', scale=1, color='#2ca02c', label=f'Vector v {tuple(v)} (at head of u)')
# Plot Resultant Vector w = u + v from origin (0, 0)
ax1.quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1, color='#d62728', label=f'Resultant u + v {tuple(w)}')

# Coordinate labels
ax1.text(u[0]/2 - 0.4, u[1]/2 + 0.3, f'u {tuple(u)}', fontsize=11, color='#1f77b4', fontweight='bold')
ax1.text(u[0] + v[0]/2 + 0.2, u[1] + v[1]/2, f'v {tuple(v)}', fontsize=11, color='#2ca02c', fontweight='bold')
ax1.text(w[0]/2 - 0.7, w[1]/2 + 0.4, f'u + v {tuple(w)}', fontsize=11, color='#d62728', fontweight='bold')

ax1.set_xlim(-1, 8)
ax1.set_ylim(-1, 9)
ax1.set_title("🔹 Head-to-Tail Method (Triangle Law)", fontsize=12, fontweight="bold")
ax1.set_xlabel("X-axis", fontsize=11)
ax1.set_ylabel("Y-axis", fontsize=11)
ax1.grid(True, linestyle="--", alpha=0.4)
ax1.axhline(0, color='black', linewidth=0.8)
ax1.axvline(0, color='black', linewidth=0.8)
ax1.legend(loc="upper left", frameon=True)

# ----------------------------------------------------------
# Subplot 2: Parallelogram Law (Both from Origin)
# ----------------------------------------------------------
ax2 = axes[1]

# Both vectors starting from origin
ax2.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, color='#1f77b4', label=f'Vector u {tuple(u)}')
ax2.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='#2ca02c', label=f'Vector v {tuple(v)}')

# Dotted lines completing the parallelogram
ax2.plot([u[0], w[0]], [u[1], w[1]], color='#2ca02c', linestyle='--', linewidth=1.8, alpha=0.7, label='Parallel to v')
ax2.plot([v[0], w[0]], [v[1], w[1]], color='#1f77b4', linestyle='--', linewidth=1.8, alpha=0.7, label='Parallel to u')

# Resultant Vector w = u + v as diagonal
ax2.quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1, color='#d62728', label=f'Diagonal: u + v {tuple(w)}')

# Component & peak markers
ax2.scatter([w[0]], [w[1]], color='#d62728', s=80, zorder=5)
ax2.text(w[0] + 0.2, w[1] - 0.2, f'({w[0]}, {w[1]})', fontsize=11, fontweight='bold', color='#d62728')

ax2.set_xlim(-1, 8)
ax2.set_ylim(-1, 9)
ax2.set_title("🔹 Parallelogram Law (Diagonal = Sum)", fontsize=12, fontweight="bold")
ax2.set_xlabel("X-axis", fontsize=11)
ax2.set_ylabel("Y-axis", fontsize=11)
ax2.grid(True, linestyle="--", alpha=0.4)
ax2.axhline(0, color='black', linewidth=0.8)
ax2.axvline(0, color='black', linewidth=0.8)
ax2.legend(loc="upper left", frameon=True)

plt.suptitle("🎯 Vector Addition Visualizations (u = [4, 2], v = [2, 5], u + v = [6, 7])", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 6️⃣ Cross Product (Vector Product)

### 📖 Concept (Simple English):
- **Cross Product ($\mathbf{u} \times \mathbf{v}$)** is defined only for **3D vectors**.
- While the Dot Product gives a single number (scalar), the **Cross Product gives a NEW 3D VECTOR**.
- **Key Geometric Property:** The resultant vector $(\mathbf{u} \times \mathbf{v})$ is **perpendicular ($90^\circ$ orthogonal)** to BOTH original vectors $\mathbf{u}$ and $\mathbf{v}$!
- **Right-Hand Rule:** If you curl the fingers of your right hand from $\mathbf{u}$ to $\mathbf{v}$, your thumb points in the direction of $\mathbf{u} \times \mathbf{v}$.

### 📐 Formulas:
1. **Magnitude & Area:**
   $$\|\mathbf{u} \times \mathbf{v}\| = \|\mathbf{u}\| \|\mathbf{v}\| \sin(\theta)$$
   - The magnitude equals the **Area of the Parallelogram** formed by $\mathbf{u}$ and $\mathbf{v}$.
   - If vectors are parallel ($\theta = 0^\circ$), their cross product is $\mathbf{0}$.

2. **Component Formula (Determinant Form):**
   $$\mathbf{u} \times \mathbf{v} = \det \begin{bmatrix} \mathbf{i} & \mathbf{j} & \mathbf{k} \\ u_1 & u_2 & u_3 \\ v_1 & v_2 & v_3 \end{bmatrix} = \begin{bmatrix} u_2 v_3 - u_3 v_2 \\ u_3 v_1 - u_1 v_3 \\ u_1 v_2 - u_2 v_1 \end{bmatrix}$$

3. **Anti-commutative Property:**
   $$\mathbf{v} \times \mathbf{u} = -(\mathbf{u} \times \mathbf{v})$$

### 🤖 Machine Learning Connection:
- **3D Computer Vision & NeRFs:** Essential for calculating **Surface Normals (perpendicular vectors)** on 3D meshes and point clouds.
- **Robotics:** Calculates torque, angular momentum, and rotational motion."
   ]
  },
  {
   "cell_type": "code",
   "execution_count": null,
   "metadata": {},
   "outputs": [],
   "source": [
    "# ==========================================================\n",
    "# 6. CROSS PRODUCT: CALCULATION, PROPERTIES & 3D VISUALIZATION\n",
    "# ==========================================================\n",
    "\n",
    "# Define two 3D vectors\n",
    "u = np.array([3, 0, 0])\n",
    "v = np.array([0, 4, 0])\n",
    "\n",
    "# 1. Calculate Cross Product using np.cross\n",
    "cross_uv = np.cross(u, v)\n",
    "cross_vu = np.cross(v, u)\n",
    "\n",
    "print(f\"Vector u                : {u}\")\n",
    "print(f\"Vector v                : {v}\")\n",
    "print(f\"Cross Product (u x v)   : {cross_uv}\")\n",
    "print(f\"Anti-commutative (v x u): {cross_vu}  (Opposite direction!)\")\n",
    "\n",
    "# 2. Orthogonality Verification (Dot product must be 0)\n",
    "dot_u = np.dot(cross_uv, u)\n",
    "dot_v = np.dot(cross_uv, v)\n",
    "print(f\"\\nCheck (u x v) · u       : {dot_u} (Perpendicular to u!)\")\n",
    "print(f\"Check (u x v) · v       : {dot_v} (Perpendicular to v!)\")\n",
    "\n",
    "# 3. Area of Parallelogram = Magnitude of Cross Product\n",
    "area_parallelogram = np.linalg.norm(cross_uv)\n",
    "print(f\"Area of Parallelogram   : {area_parallelogram:.2f} sq units\")\n",
    "\n",
    "# ==========================================================\n",
    "# 3D VISUALIZATION OF CROSS PRODUCT\n",
    "# ==========================================================\n",
    "fig = plt.figure(figsize=(10, 8))\n",
    "ax = fig.add_subplot(111, projection='3d')\n",
    "\n",
    "origin = np.array([0, 0, 0])\n",
    "\n",
    "# Plot Vector u (along X-axis)\n",
    "ax.quiver(origin[0], origin[1], origin[2], u[0], u[1], u[2],\n",
    "          color='#1f77b4', arrow_length_ratio=0.15, linewidth=3.5, label=f'Vector u {tuple(u)}')\n",
    "\n",
    "# Plot Vector v (along Y-axis)\n",
    "ax.quiver(origin[0], origin[1], origin[2], v[0], v[1], v[2],\n",
    "          color='#2ca02c', arrow_length_ratio=0.15, linewidth=3.5, label=f'Vector v {tuple(v)}')\n",
    "\n",
    "# Plot Resultant Vector u x v (perpendicular along Z-axis)\n",
    "ax.quiver(origin[0], origin[1], origin[2], cross_uv[0], cross_uv[1], cross_uv[2],\n",
    "          color='#d62728', arrow_length_ratio=0.12, linewidth=4, label=f'Cross Product u x v {tuple(cross_uv)}')\n",
    "\n",
    "# Draw the parallelogram formed by u and v on the XY-plane\n",
    "corner = u + v\n",
    "ax.plot([u[0], corner[0]], [u[1], corner[1]], [u[2], corner[2]], color='#2ca02c', linestyle='--', alpha=0.6)\n",
    "ax.plot([v[0], corner[0]], [v[1], corner[1]], [v[2], corner[2]], color='#1f77b4', linestyle='--', alpha=0.6)\n",
    "\n",
    "# Text annotations for each vector tip\n",
    "ax.text(u[0] + 0.2, u[1], u[2], f'u {tuple(u)}', fontsize=11, color='#1f77b4', fontweight='bold')\n",
    "ax.text(v[0], v[1] + 0.3, v[2], f'v {tuple(v)}', fontsize=11, color='#2ca02c', fontweight='bold')\n",
    "ax.text(cross_uv[0], cross_uv[1], cross_uv[2] + 0.5, f'u x v {tuple(cross_uv)}\\n(90° to both u & v)',\n",
    "        fontsize=11, color='#d62728', fontweight='bold')\n",
    "\n",
    "# Axes settings & viewing angle\n",
    "ax.set_xlim([-1, 5])\n",
    "ax.set_ylim([-1, 5])\n",
    "ax.set_zlim([-1, 14])\n",
    "\n",
    "ax.set_xlabel('X-axis', fontweight='bold')\n",
    "ax.set_ylabel('Y-axis', fontweight='bold')\n",
    "ax.set_zlabel('Z-axis (Perpendicular)', fontweight='bold')\n",
    "ax.set_title('⚡ 3D Cross Product (u x v): Always Perpendicular to Both Vectors!', fontsize=13, fontweight='bold')\n",
    "ax.legend(loc='upper left')\n",
    "ax.view_init(elev=20, azim=35)\n",
    "\n",
    "plt.tight_layout()\n",
    "plt.show()\n"
   ]
  },
  {
   "cell_type": "markdown",
   "metadata": {},
   "source": [
    "## 7️⃣ Linear Algebra Foundations (Vectors, Dot Product, Matrix Operations)

### 📖 Concept (Simple English):
- **Scalar (0D):** A single numerical value.
- **Vector (1D):** An array of numbers with magnitude and direction.
- **Matrix (2D):** A rectangular table of numbers with rows and columns.
- **Dot Product:** Measures similarity/projection between two vectors (Dot product $= 0$ means orthogonal/perpendicular!).
- **Matrix Inverse ($A^{-1}$):** A matrix such that $A \cdot A^{-1} = I$ (Identity Matrix).
```

```python
# ==========================================================
# 7. VECTORS, DOT PRODUCT & MATRIX MULTIPLICATION
# ==========================================================

# --- Vectors & Dot Product ---
u = np.array([3, 4])
v = np.array([4, -3])

dot_prod = np.dot(u, v)
print(f"Vector u: {u}")
print(f"Vector v: {v}")
print(f"Dot Product u · v: {dot_prod}")
if dot_prod == 0:
    print("✅ u and v are ORTHOGONAL (Perpendicular) vectors because u · v = 0!")

# --- Matrices & Inverse ---
A = np.array([[2, 1],
              [1, 3]])

det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)

print("\nMatrix A:")
print(A)
print(f"Determinant of A: {det_A:.2f}")
print("Inverse Matrix A^(-1):")
print(inv_A)

# Verification: A @ A_inv = Identity Matrix I
I = np.round(A @ inv_A)
print("Verification (A @ A^(-1) = Identity):")
print(I)
```
