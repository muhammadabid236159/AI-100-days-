## What is File I/O?

**File I/O = File Input and Output**

- **Input (I):** Reading data from a file
- **Output (O):** Writing data into a file

Files allow programs to store data permanently!

## Why Use Files?

✅ Save data permanently
✅ Read previously saved data
✅ Store records (students, employees, products)
✅ Share data between programs
✅ Backup important information

## Types of Files

### 1. Text Files (.txt)
- Human-readable text
- Example: student.txt

### 2. Binary Files (.bin, .jpg, .pdf)
- Data in binary format (0s and 1s)
- Images, videos, audio, PDFs, programs

## File Modes

| Mode | Meaning |
|------|---------|
| 'r' | Read (default) |
| 'w' | Write (overwrites existing) |
| 'a' | Append (add at end) |
| 'x' | Create new (error if exists) |
| 'r+' | Read and write |
| 'w+' | Read and write (overwrites) |
| 'a+' | Read and append |

## Reading Files

### Methods:
- `read()` - Read entire file
- `readline()` - Read one line
- `readlines()` - Read all lines as list

## Best Practice: Using 'with' Statement

The `with` statement automatically closes the file.
This is RECOMMENDED!

```python
# Example: Writing to a file
with open("student.txt", "w") as file:
    file.write("Name: Muhammad Abid\n")
    file.write("Roll No: 236159\n")
    file.write("Marks: 90\n")

print("Data written successfully!")

# Example: Reading from a file
with open("student.txt", "r") as file:
    content = file.read()
    print("File Content:")
    print(content)

# Example: Appending to a file
with open("student.txt", "a") as file:
    file.write("Grade: A\n")

print("Data appended successfully!")
```

## Advantages & Disadvantages

### Advantages:
✅ Permanent data storage
✅ Easy to read and update
✅ Useful for logs, reports, databases
✅ Makes programs more practical

### Disadvantages:
❌ Slower than memory operations
❌ Requires error handling
❌ Large files consume memory
