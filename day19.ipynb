```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara"],
    "Math": [80, 75, 90],
    "English": [70, 85, 95]
})

print(df)
```

```python
melted=df.melt(id_vars=["Name"], var_name="Subject", value_name="Score")
print(melted)
```

```python
pivoted = melted.pivot(
    index="Name",
    columns="Subject",
    values="Score"
)
print(pivoted)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ali", "Ahmed", "Sara", "Usman", "Fatima"],
    "Jan": [100, 200, 150, 180, 220],
    "Feb": [120, 180, 170, 190, 210],
    "Mar": [150, 220, 160, 200, 230]
})

print(df)
```

```python
melted = df.melt(
    id_vars="Name",
    var_name="Month",
    value_name="Sales"
)
melted
```

```python
pivoted = melted.pivot(
    index="Name",
    columns="Month",
    values="Sales"
)

print(pivoted)
```
