## Abstraction in OOP

Abstraction means showing only essential information while hiding implementation details.

**Real-Life Example:**
When using an ATM:
- You enter PIN, amount
- You get cash
- You DON'T see internal transaction processing

This is Abstraction!

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

class Dog(Animal):
    def make_sound(self):
        return "dog bark"

class Cat(Animal):
    def make_sound(self):
        return "cat meow"

dog = Dog()
cat = Cat()

print(dog.make_sound())
print(cat.make_sound())
```

## Real-World Example: Employee Salary Calculation

Different employee types calculate salary differently

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    @abstractmethod
    def calculate_salary(self, days, amount):
        pass

class FullTimeEmployee(Employee):
    def calculate_salary(self, days, amount):
        fulltime_salary = days * amount
        return fulltime_salary

class PartTimeEmployee(Employee):
    def calculate_salary(self, days, amount):
        parttime_salary = days * (amount / 2)
        return parttime_salary

full = FullTimeEmployee()
part = PartTimeEmployee()

print(f"Full Time Employee Salary: {full.calculate_salary(30, 2000)}")
print(f"Part Time Employee Salary: {part.calculate_salary(30, 2000)}")
```

## Project: Food Ordering App

Abstract food class with different recipes

```python
from abc import ABC, abstractmethod

class Food(ABC):
    @abstractmethod
    def prepare(self):
        pass

class Pizza(Food):
    def prepare(self):
        return "Preparing pizza..."
    
class Burger(Food):
    def prepare(self):
        return "Preparing burger..."
    
customer1 = Pizza()
print(customer1.prepare())

customer2 = Burger()
print(customer2.prepare())
```

## Polymorphism

Polymorphism = "One interface, many forms"

Same method name, different implementations in different classes

```python
class Dog:
    def sound(self):
        return "Dog Barks"

class Cat:
    def sound(self):
        return "Cat Meows"

class Cow:
    def sound(self):
        return "Cow Moos"

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(animal.sound())
```

## Vehicle System (Polymorphism Example)

Different vehicles start differently

```python
class Vehicle:
    def start(self):
        return "start the vehicle"
    
class Bike(Vehicle):
    def start(self):
        return "bike has started"

class Car(Vehicle):
    def start(self):
        return "car has started"

vehicles = [Vehicle(), Bike(), Car()]
for vehicle in vehicles:
    print(vehicle.start())
```
