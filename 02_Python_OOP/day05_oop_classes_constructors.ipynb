## Introduction to Object-Oriented Programming (OOP)

OOP is a programming paradigm based on the concept of objects, which contain data (attributes) and methods (functions).

```python
# Simple Student class
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def show(self):
        print(f"My name is {self.name} and I am {self.age} years old.")

# Create objects
s1 = Student("abid", 19)
s2 = Student("abbas", 23)
s1.show()
s2.show()
```

## Types of Constructors

In Python, __init__() is the constructor that initializes objects when created.

```python
class Car:
    def __init__(self, name, brand):
        self.name = name
        self.brand = brand

    def is_active(self, active):
        self.active = active

    def show(self):
        print(f"Car Name: {self.name}")
        print(f"Brand: {self.brand}")
        print(f"Active: {self.active}")

# Create and use
c1 = Car("Prius", "Toyota")
c1.is_active(True)
c1.show()
```

## Class Variables vs Instance Variables

- **Class Variables:** Shared across all instances
- **Instance Variables:** Unique to each object

```python
class University:
    department_name = "Computer Science"  # Class variable

    def __init__(self, name, grade, cgpa):
        self.name = name  # Instance variables
        self.grade = grade
        self.cgpa = cgpa
    
    def show_information(self):
        return (University.department_name, self.name, self.grade, self.cgpa)
    
student1 = University("abid", "A", 3.58)
student2 = University("abbas", "B", 3.2)

print(student1.show_information())
print(student2.show_information())
```

## @classmethod and @staticmethod

- **@classmethod:** Methods that work with class variables
- **@staticmethod:** Independent functions within a class

```python
class CompanyLaptop:
    storage = "SSD 256GB"  # Class variable

    def __init__(self, ram, brand):
        self.ram = ram
        self.brand = brand

    def show(self):
        print(f"Brand: {self.brand}, RAM: {self.ram}GB")

    @classmethod
    def show_storage(cls):
        print(cls.storage)
    
    @staticmethod
    def calculate_price(price, quantity):
        final_price = price * quantity
        discount = final_price * 10 / 100
        payable = final_price - discount
        return final_price, discount, payable

# Usage
laptop1 = CompanyLaptop(16, "Dell")
laptop1.show()
laptop1.show_storage()
final, discount, payable = laptop1.calculate_price(10000, 5)
print(f"Final Price: {final}, Discount: {discount}, Payable: {payable}")
```
